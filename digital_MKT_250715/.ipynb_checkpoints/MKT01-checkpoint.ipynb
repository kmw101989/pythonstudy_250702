```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup as bs 
import time 
from openpyxl import Workbook

#1.셀레니움 기본 옵션 설정
options = Options()
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-sfm-usage")
options.add_argument("--headless")

driver = webdriver.Chrome(options = options)

# 2.페이지 접속 
url = "https://www.kurly.com/goods/1000830585"
driver.get(url)

time.sleep(3) #페이지 로딩 대기시간 

# 3. HTML 가져오기 
soup = bs(driver.page_source,"html.parser")

# 4. 리뷰 텍스트 추출하기 
data = soup.select(".css-y49dcn")
data_lists = [item.get_text().strip() for item in data]



driver.quit

wb = Workbook() 
ws = wb.active
ws.title = "Kurly Reviews"

ws.append(["리뷰내용"])
for item in data_lists :
    ws.append([item])

wb.save("kurly_reviews.xlsx")
print("엑셀 저장 완료 :kurly_reviews.xlsx")
```
