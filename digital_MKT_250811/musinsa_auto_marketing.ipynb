```python
from konlpy.tag import Okt
from oauth2client.service_account import ServiceAccountCredentials
import gspread
import gspread.exceptions as gse
import pandas as pd
import time

# 1) 데이터 준비
file = "musinsa_top30_review.xlsx"
df = pd.read_excel(file)

okt = Okt()
df["리뷰_키워드"] = df["리뷰"].apply(lambda x: ", ".join(okt.nouns(str(x))))

# 2) 인증
scope = [
    "https://www.googleapis.com/auth/spreadsheets",
    "https://www.googleapis.com/auth/drive",
]
creds = ServiceAccountCredentials.from_json_keyfile_name(
    "pure-chariot-468703-k8-5ad1fe73f7f0.json", scope
)
client = gspread.authorize(creds)

# 3) 스프레드시트 열기
spreadsheet = client.open("무신사 리뷰 분석")

# 4) 첫 시트 업데이트(원본)
sheet1 = spreadsheet.sheet1
sheet1.clear()
sheet1.update([df.columns.values.tolist()] + df.values.tolist())

# 5) 안전한 get_or_create
def get_or_create_worksheet(ss, title, rows=1000, cols=20, retries=3, delay=0.8):
    for _ in range(retries):
        try:
            return ss.worksheet(title)
        except gse.WorksheetNotFound:
            # 없으면 생성 시도
            try:
                ss.add_worksheet(title=title, rows=rows, cols=cols)  # <-- cols 사용!
            except gse.APIError as e:
                # 이미 존재하지만 전파 지연/경쟁 상태일 수도 있음 -> 무시하고 재시도
                pass
            time.sleep(delay)
    # 마지막 재시도
    return ss.worksheet(title)

# 6) 키워드 분석 시트 갱신
sheet2 = get_or_create_worksheet(spreadsheet, "키워드 분석", rows=2000, cols=30)
sheet2.clear()
sheet2.update(
    [["상품명", "리뷰", "리뷰_키워드"]]
    + df[["상품명", "리뷰", "리뷰_키워드"]].values.tolist()
)

print("스프레드시트에 원복+키워드 분석 자료 업데이트 완료")

```
