```python

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pyautogui
import pickle
import pymysql 

# db = pymysql.connect(
#     host = "localhost",
#     port = 3306,
#     user = "root",
#     passwd = "hithere4961",
#     db = "ecommerce_v2",
#     charset = "utf8"
# ) 


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("--start-maximized")
options.add_argument("--lang=ko_KR")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")


driver = webdriver.Chrome(service=service, options=options)

url = "https://tickets.interpark.com/contents/ranking?genre=LEISURE"
driver.get(url)
time.sleep(2)

buttons = driver.find_elements(By.CSS_SELECTOR, "div.RadioButton_groupWrap__2bab4")
for button in buttons : 
    button.click()
    time.sleep(2)
    items = driver.find_elements(By.CSS_SELECTOR,'div.responsive-ranking-list_rankingItemInner__mMLxe div.responsive-ranking-list_imageWrap__ROlYW img')
    for item in items :
        # ✅ src에서 마지막 '/' 이후부터 '_' 전까지(예: 25009974) 추출
        src = item.get_attribute("src") or item.get_attribute("data-src") or item.get_attribute("currentSrc")
        if not src:
            continue
        start = src.rfind('/') + 1
        end = src.find('_', start)
        pid = src[start:end] if end != -1 else src[start:start+8]  # fallback
        print(pid)
    
       

time.sleep(2)
# driver.quit()
```
