```python
#네이버 지식백과에서 '콘텐츠 마케팅' 항목의 본문 요약을 추출하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://terms.naver.com/entry.naver?docId=2275742&cid=42266&categoryId=51204"
driver.get(url)

elements = driver.find_elements(By.CSS_SELECTOR, "p.txt")
for element in elements: 
    if element.text.strip():
        print(element.text)


time.sleep(1)

driver.quit()
```

```python
#한국관광공사 '추천 여행지' 페이지에서 지역별 여행지명을 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
import pyautogui

service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://korean.visitkorea.or.kr/main/area.do"
driver.get(url)
time.sleep(2)

all_list = {}
for i in range(17):
    pyautogui.click(x=1030 , y=320)
    time.sleep(2)
    titles = driver.find_elements(By.CSS_SELECTOR ,"ul > li > a > strong")
    city = driver.find_element(By.XPATH, "//*[@id='contents']/div[3]/div/p/strong/em").text
    title_list = [title.text.strip() for title in titles if title.text.strip()]
    if city in all_list:
        all_list[city].extend(title_list)
    else:
        all_list[city] = title_list

for city, places in all_list.items():
    print(f"[{city}]")
    for p in places:
        print(f" - {p}")
    print()
    
driver.quit()
```

```python


#교보문고 경제 분야 주간 베스트셀러 도서 제목 수집하세요.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://product.kyobobook.co.kr/category/KOR/1309#?page=1&type=best"
driver.get(url)

elements = driver.find_elements(By.CSS_SELECTOR, "span.prod_name")
for element in elements: 
    if element.text.strip():
        print(element.text)


time.sleep(1)

driver.quit()
```

```python


#멜론 차트 1~10위 노래 제목과 가수를 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://www.melon.com/chart/"
driver.get(url)

info = []
    

titles = driver.find_elements(By.XPATH, "//a[contains(@href , 'play')]")
artists = driver.find_elements(By.CSS_SELECTOR, "div.ellipsis.rank02")

for idx, (title, artist) in enumerate(zip(titles, artists)):
    if idx >= 10:
        break
    data = {
        "title": title.text.strip(),
        "artist": artist.text.strip()
    }
    info.append(data)

for item in info:
    print(f"{item['title']} - {item['artist']}")

time.sleep(1)

driver.quit()
```
