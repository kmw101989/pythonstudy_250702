```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as ec 
from webdriver_manager.chrome import ChromeDriverManager
import time 
import pandas as pd 




options = Options()
# options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--start-maximized")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--lang=ko_KR")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service = service  , options = options)

url = "https://tickets.interpark.com/contents/genre/concert"

height=2500
driver.get(url)
time.sleep(3)
idx = 1 
data = [] 
titles = []
places= []
dates = []
for i in range(5) : 
    driver.execute_script(f"window.scrollTo(0,{height})")
    height += 2500
    time.sleep(1)
    elements = driver.find_elements(By.XPATH , "//ul[contains(@class, 'TicketItem_contentsWrap__xCe3A')]")
    for element in elements :
        title = element.find_elements(By.XPATH, ".//li")[0].text
        place = element.find_elements(By.XPATH, ".//li")[1].text
        date = element.find_elements(By.XPATH, ".//li")[2].text
        # print(f"{idx}: {title}\n{date}\n")
        idx += 1 
        titles.append(title)
        places.append(place)
        dates.append(date)

time.sleep(1)
driver.quit()

data = []
seen_title_date = set()

for t, d, p in zip(titles, dates, places):
    key = (t.strip(), d.strip())
    if key not in seen_title_date:
        seen_title_date.add(key)
        data.append({
            "title": t.strip(),
            "date": d.strip(),
            "place": p.strip()
        })


df = pd.DataFrame(data)
df.head(100)
```

```python
#마이리얼트립에서 인기 도시별 여행 상품 제목을 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.action_chains import ActionChains
import time
import pyautogui


service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://www.myrealtrip.com/"
driver.get(url)

time.sleep(1)
pyautogui.click(x= 850, y=850)
time.sleep(1)

swipe = driver.find_element(By.CSS_SELECTOR, "div.css-15thzkg")
actions = ActionChains(driver)
fav_button = driver.find_element(By.CSS_SELECTOR, "div.css-1f9jvpk")
time.sleep(1)
titles = set()
city_product_dict = {}
city_name = fav_button.text.strip()
buttons_city = driver.find_elements(By.CSS_SELECTOR ,"button.css-8nly55")

for i in range(8) :
    fav_button.click()
    button = buttons_city = driver.find_elements(By.CSS_SELECTOR ,"button.css-8nly55")[i]
    button.click()
    time.sleep(1)
    
    cities1= driver.find_elements(By.CSS_SELECTOR, "span.css-43qqj2")
    for city1 in cities1:
        if city1.text.strip():
            titles.add(city1.text.strip())
    actions.click_and_hold(swipe).move_by_offset(-800, 0).release().perform()
    time.sleep(1)
    
    cities2 = driver.find_elements(By.CSS_SELECTOR, "span.css-43qqj2")
    for city2 in cities2:
        if city2.text.strip():
            titles.add(city1.text.strip())
  

    city_product_dict[city_name] = list(titles)
    time.sleep(1)

# for title in titles :
#     print(title)
for city, products in city_product_dict.items():
    print(f"\n📍 {city}")
    for title in products:
        print(f" - {title}")
driver.quit()


```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from webdriver_manager.chrome import ChromeDriverManager
import time
import pyautogui

# 브라우저 세팅
service = Service(ChromeDriverManager().install())
options = Options()
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

# 사이트 접속
url = "https://www.myrealtrip.com/"
driver.get(url)

time.sleep(1)
pyautogui.click(x=850, y=850)  # 초기 팝업 닫기
time.sleep(1)

# 인기 도시 영역 준비
fav_button = driver.find_element(By.CSS_SELECTOR, "div.css-1f9jvpk")
fav_button.click()
time.sleep(2)

swipe = driver.find_element(By.CSS_SELECTOR, "div.css-15thzkg")
actions = ActionChains(driver)

# 도시별 상품 저장
city_product_dict = {}

# 도시 버튼들 탐색
buttons_city = driver.find_elements(By.CSS_SELECTOR, "button.css-8nly55")
num_cities = min(8, len(buttons_city))

for i in range(num_cities):
    fav_button.click()
    time.sleep(1)

    buttons_city = driver.find_elements(By.CSS_SELECTOR, "button.css-8nly55")
    button = buttons_city[i]

    city_name = button.text.strip()
    if not city_name:
        continue

    button.click()
    time.sleep(2)

    # 해당 도시의 상품 제목들을 담을 set
    titles = set()

    # 1차 상품 수집
    cities1 = driver.find_elements(By.CSS_SELECTOR, "span.css-43qqj2")
    for city1 in cities1:
        title = city1.text.strip()
        if title:
            titles.add(title)

    # 스와이프 후 수집
    actions.click_and_hold(swipe).move_by_offset(-800, 0).release().perform()
    time.sleep(2)

    cities2 = driver.find_elements(By.CSS_SELECTOR, "span.css-43qqj2")
    for city2 in cities2:
        title = city2.text.strip()
        if title:
            titles.add(title)

    # 딕셔너리에 저장
    city_product_dict[city_name] = list(titles)
    time.sleep(1)

# 출력
for city, products in city_product_dict.items():
    print(f"\n📍 {city}")
    for idx ,title in enumerate(products,1):
        print(f"{idx} - {title}")

driver.quit()
```
