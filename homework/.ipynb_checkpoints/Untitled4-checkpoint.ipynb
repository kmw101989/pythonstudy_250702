```python
#G마켓에서 '게이밍 키보드' 검색 후 스크롤을 내려 상품명과 리뷰수를 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://www.gmarket.co.kr/n/search?keyword=%EA%B2%8C%EC%9D%B4%EB%B0%8D+%ED%82%A4%EB%B3%B4%EB%93%9C"
driver.get(url)

elements = driver.find_elements(By.CSS_SELECTOR, "div.box__component-itemcard div.box__information")
for idx, element in enumerate(elements,1) : 
    try:
        prodname = element.find_element(By.CSS_SELECTOR, "span.text__item").get_attribute("title")
        rvcount = element.find_element(By.CSS_SELECTOR, "ul > li.list-item.list-item__feedback-count > span.text").text
        print(f"{idx} : {prodname}\n리뷰:{rvcount}개")
    except:
        pass
time.sleep(1)

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

# 드라이버 설정
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

url = "https://www.gsshop.com/shop/tv/tvScheduleMain.gs?lseq=415680-1&gsid=ECmain-AU415680-AU415680-1"
driver.get(url)
time.sleep(3)

# ✅ 시간대별 상품명 저장 딕셔너리
times = {}

elements = driver.find_elements(By.CSS_SELECTOR, "article.items")
for element in elements:
    try:
        # 시간 텍스트 추출
        time_key = element.find_element(By.CSS_SELECTOR, "span.times").text.strip()

        # 상품명 리스트 추출
        products = element.find_elements(By.CSS_SELECTOR, "a")
        for product in products:
            if time_key not in times:
                times[time_key] = []  # 리스트 초기화
            times[time_key].append(product.text.strip()) 

    except Exception as e:
        continue  # 오류 발생 시 건너뜀

time.sleep(2)
driver.quit()

# ✅ 결과 출력
for time_slot, products in times.items():
    print(f"{time_slot} ▶ {products}\n")
```
