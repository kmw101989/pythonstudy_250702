```python
# -*- coding: utf-8 -*-
# 쿠팡 주문/배송 조회: 전용 크롬 프로필 + 로그인 성공 신호 대기 + 콘솔 출력(저장 X)

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time, os, sys

# ========= 사용자 설정 =========
CHROMEDRIVER = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"

# 전용 프로필(일반 크롬과 절대 겹치지 않는 경로!)
USER_DATA_DIR = r"C:\SeleniumProfiles\UserData"
PROFILE_DIR   = "Profile_Auto"

# 주문내역 페이지
ORDER_URL = "https://mc.coupang.com/ssr/desktop/order/list?requestYear=2024"

# 네가 제공한 셀렉터(필요 시 DOM 보고 미세 조정)
ORDER_CARD_SEL = "div.sc-fimazj-0.gKYVxm"                              # 주문내역 제품 블럭
NAME_IN_CARD   = "span.sc-755zt3-1.sc-8q24ha-1.ksjYDU.ifMZxv"          # 제품명
STATUS_IN_CARD = "span.sc-755zt3-0.anmor"                               # 배송상태

# 해시 변동 대비용 폴백
NAME_FALLBACKS = [
    "span.sc-755zt3-1.sc-8q24ha-1",
    "span[class*='sc-8q24ha-1']",
    "span[class*='sc-755zt3-1']",
]
STATUS_FALLBACKS = [
    "span.sc-755zt3-0.anmor",
    "span[class*='sc-755zt3-0']",
]

STATUS_KEYWORDS = ["배송중", "배송완료", "상품준비중", "결제완료", "구매확정", "취소완료", "교환진행", "반품진행"]


# ========= 드라이버 생성 =========
def build_driver():
    service = Service(CHROMEDRIVER)
    options = Options()

    # 전용 프로필(폴더 보장)
    if os.path.isfile(USER_DATA_DIR):
        os.remove(USER_DATA_DIR)
    os.makedirs(USER_DATA_DIR, exist_ok=True)
    options.add_argument(f"--user-data-dir={USER_DATA_DIR}")
    options.add_argument(f"--profile-directory={PROFILE_DIR}")

    # 기본 옵션
    options.add_argument("--no-first-run")
    options.add_argument("--no-default-browser-check")
    options.add_argument("--disable-extensions")
    options.add_argument("--start-maximized")
    options.add_argument("--lang=ko_KR")
    options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")

    # 자동화 흔적 최소화
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option("useAutomationExtension", False)

    driver = webdriver.Chrome(service=service, options=options)

    # navigator.webdriver 감춤
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined});"
    })
    return driver


# ========= 상태 감지 =========
def is_logged_in(driver):
    try:
        cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
        return any(card.is_displayed() for card in cards)
    except Exception:
        return False

def is_captcha_visible(driver):
    try:
        for css in ["._loginCaptchaContainer", "._loginCaptchaImage", "._loginCaptchaInput",
                    "#captcha", ".captcha", "[class*='captcha']"]:
            for el in driver.find_elements(By.CSS_SELECTOR, css):
                if el.is_displayed():
                    return True
        return False
    except Exception:
        return False

def wait_until_logged_in(driver, timeout=30):
    print("🔐 수동 로그인 대기 중…(주문 목록이 보이면 자동 진행)")
    start = time.time()
    while time.time() - start < timeout:
        if is_logged_in(driver):
            print("✅ 로그인 성공 감지")
            return True
        if is_captcha_visible(driver):
            print("⚠️ 캡차/추가 인증 감지 → 화면에서 직접 처리하세요. (스크립트는 대기)")
        time.sleep(1)
    print("⏰ 로그인 대기 타임아웃")
    return False


# ========= 세션 보장(프로필 기반) =========
def ensure_session(driver):
    """
    전용 프로필 기반으로:
    - 로그인 상태면 곧장 진행
    - 로그인 페이지면 수동 로그인 성공 신호 대기
    """
    driver.get(ORDER_URL)
    time.sleep(2)

    if is_logged_in(driver):
        print("ℹ️ 이미 로그인 상태로 확인")
        return

    if "login" in driver.current_url.lower() or "member" in driver.current_url.lower():
        ok = wait_until_logged_in(driver)
        if not ok:
            print("❌ 로그인 실패. 인증/캡차를 마치고 다시 실행하세요.")
            sys.exit(1)
        return

    # 그 외 애매한 경우: 약간 더 기다렸다가 판단
    time.sleep(2)
    if not is_logged_in(driver):
        print("❓ 로그인 상태 확인 불가. 페이지를 한 번 새로고침합니다.")
        driver.get(ORDER_URL)
        time.sleep(2)
        if not is_logged_in(driver):
            print("❌ 로그인 필요. 화면에서 로그인 후 다시 실행하세요.")
            sys.exit(1)


# ========= 스크랩 & 출력 =========
def scrape_and_print(driver):
    wait = WebDriverWait(driver, 10)
    wait.until(lambda d: d.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL))
    time.sleep(0.8)

    cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
    rows = []

    for card in cards:
        # 배송 상태 (카드 기준 1개 가정)
        status_text = ""
        try:
            status_el = card.find_element(By.CSS_SELECTOR, STATUS_IN_CARD)
            status_text = status_el.text.strip()
        except Exception:
            for css in STATUS_FALLBACKS:
                try:
                    status_el = card.find_element(By.CSS_SELECTOR, css)
                    status_text = status_el.text.strip()
                    if status_text:
                        break
                except Exception:
                    pass
        if not status_text:
            blob = (card.text or "")
            for kw in STATUS_KEYWORDS:
                if kw in blob:
                    status_text = kw
                    break

        # 제품명 (복수 가능)
        name_els = card.find_elements(By.CSS_SELECTOR, NAME_IN_CARD)
        if not name_els:
            for css in NAME_FALLBACKS:
                name_els = card.find_elements(By.CSS_SELECTOR, css)
                if name_els:
                    break

        for el in name_els:
            name = (el.text or el.get_attribute("title") or el.get_attribute("aria-label") or "").strip()
            if name:
                rows.append((name, status_text))

    # 콘솔 출력만
    print(f"📦 수집 건수: {len(rows)}")
    for i, (n, s) in enumerate(rows, 1):
        print(f"{i}. {n} | {s}")


# ========= 실행 =========
if __name__ == "__main__":
    # 일반 크롬과 동시 실행/여러 인스턴스 동시 실행 금지!
    driver = build_driver()
    try:
        ensure_session(driver)
        scrape_and_print(driver)
        time.sleep(1)
    finally:
        driver.quit()
```

```python
# -*- coding: utf-8 -*-
"""
쿠팡 주문내역 스크랩: 우회 장치 Ablation 테스트 프레임워크
- SCENARIO에 따라 우회 장치들을 하나씩 끄고 검증
- 로그인 대기: 30초 / 요소 대기: 10초
- 저장(X), 콘솔 출력(O)
"""

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time, os, sys

# ===== 사용자 경로 =====
CHROMEDRIVER = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"
BASE_USER_DATA_ROOT = r"C:\SeleniumProfiles\UserData"  # 전용 프로필 루트

# ===== 대상 페이지 & 셀렉터 =====
ORDER_URL = "https://mc.coupang.com/ssr/desktop/order/list?requestYear=2024"

ORDER_CARD_SEL = "div.sc-fimazj-0.gKYVxm"                              # 주문내역 제품 블럭
NAME_IN_CARD   = "span.sc-755zt3-1.sc-8q24ha-1.ksjYDU.ifMZxv"          # 제품명
STATUS_IN_CARD = "span.sc-755zt3-0.anmor"                               # 배송상태
NAME_FALLBACKS = [
    "span.sc-755zt3-1.sc-8q24ha-1",
    "span[class*='sc-8q24ha-1']",
    "span[class*='sc-755zt3-1']",
]
STATUS_FALLBACKS = [
    "span.sc-755zt3-0.anmor",
    "span[class*='sc-755zt3-0']",
]
STATUS_KEYWORDS = ["배송중", "배송완료", "상품준비중", "결제완료", "구매확정", "취소완료", "교환진행", "반품진행"]

# ===== 시나리오 선택 =====
# 'baseline'                     : 모든 우회 ON
# 'no_exclude_switches'          : excludeSwitches/useAutomationExtension OFF
# 'no_hide_webdriver'            : navigator.webdriver 숨김 OFF
# 'no_ua_lang'                   : UA/언어 플래그 OFF
# 'no_profile'                   : 전용 프로필 OFF(=> 충돌/로그인 필요 가능)
# 'no_login_marker_wait'         : 주문카드 등장 대기 OFF(즉시 스크랩 시도)
# 'no_captcha_detect'            : 캡차 감지 OFF(메시지 안내 없음)
SCENARIO = 'baseline'

# ===== 시나리오별 토글 맵 =====
TOGGLES_BY_SCENARIO = {
    'baseline': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': True,
        'UA_LANG': True,
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': True,
    },
    'no_exclude_switches': {
        'EXCLUDE_SWITCHES': False,  # 끔
        'HIDE_WEBDRIVER': True,
        'UA_LANG': True,
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': True,
    },
    'no_hide_webdriver': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': False,    # 끔
        'UA_LANG': True,
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': True,
    },
    'no_ua_lang': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': True,
        'UA_LANG': False,           # 끔
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': True,
    },
    'no_profile': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': True,
        'UA_LANG': True,
        'USE_PROFILE': False,       # 끔
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': True,
    },
    'no_login_marker_wait': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': True,
        'UA_LANG': True,
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': False, # 끔
        'CAPTCHA_DETECT': True,
    },
    'no_captcha_detect': {
        'EXCLUDE_SWITCHES': True,
        'HIDE_WEBDRIVER': True,
        'UA_LANG': True,
        'USE_PROFILE': True,
        'WAIT_LOGIN_MARKER': True,
        'CAPTCHA_DETECT': False,    # 끔
    },
}

# ===== 타임아웃(요청대로 축소) =====
LOGIN_WAIT_SEC = 30   # 로그인 성공 대기
ELEMENT_WAIT_SEC = 10 # 요소 로딩 대기

def build_driver(toggles, profile_tag):
    service = Service(CHROMEDRIVER)
    options = Options()

    # 전용 프로필 ON이면, 루트 아래에 시나리오별 하위 프로필을 둬서 상호 독립 실행
    # 예: C:\SeleniumProfiles\UserData\baseline , \no_exclude_switches ...
    if toggles['USE_PROFILE']:
        user_data_dir = os.path.join(BASE_USER_DATA_ROOT, profile_tag)
        os.makedirs(user_data_dir, exist_ok=True)
        options.add_argument(f"--user-data-dir={user_data_dir}")
        options.add_argument(f"--profile-directory=Profile_Auto")

    # 기본 옵션
    options.add_argument("--no-first-run")
    options.add_argument("--no-default-browser-check")
    options.add_argument("--disable-extensions")
    options.add_argument("--start-maximized")
    if toggles['UA_LANG']:
        options.add_argument("--lang=ko_KR")
        options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")

    if toggles['EXCLUDE_SWITCHES']:
        options.add_experimental_option("excludeSwitches", ["enable-automation"])
        options.add_experimental_option("useAutomationExtension", False)

    driver = webdriver.Chrome(service=service, options=options)

    if toggles['HIDE_WEBDRIVER']:
        driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
            "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined});"
        })

    return driver

def captcha_visible(driver):
    if not TOGGLES['CAPTCHA_DETECT']:
        return False
    try:
        for css in ["._loginCaptchaContainer", "._loginCaptchaImage", "._loginCaptchaInput",
                    "#captcha", ".captcha", "[class*='captcha']"]:
            els = driver.find_elements(By.CSS_SELECTOR, css)
            if any(e.is_displayed() for e in els):
                return True
        return False
    except Exception:
        return False

def logged_in(driver):
    try:
        cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
        return any(c.is_displayed() for c in cards)
    except Exception:
        return False

def ensure_session(driver):
    driver.get(ORDER_URL)
    time.sleep(1.5)

    if logged_in(driver):
        print("ℹ️ 이미 로그인 상태")
        return True

    # 로그인 대기 OFF면 즉시 진행(실패 가능성 높음)
    if not TOGGLES['WAIT_LOGIN_MARKER']:
        print("⚠️ 로그인 성공 신호 대기 스킵")
        return True

    # 로그인 대기 ON: 30초 동안 주문카드 등장 기다리기
    print("🔐 수동 로그인 대기… (최대 30초)")
    start = time.time()
    while time.time() - start < LOGIN_WAIT_SEC:
        if logged_in(driver):
            print("✅ 로그인 성공 감지")
            return True
        if captcha_visible(driver):
            print("⚠️ 캡차/추가인증 감지 → 수동 처리해주세요(대기 중)")
        time.sleep(1)

    print("⏰ 로그인 대기 타임아웃")
    return False

def scrape_and_print(driver):
    # 요소 대기 10초
    wait = WebDriverWait(driver, ELEMENT_WAIT_SEC)
    # 대기 스킵 모드일 때도 최소한 존재 확인 한번은 시도
    try:
        wait.until(lambda d: d.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL))
    except Exception:
        pass
    time.sleep(0.5)

    cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
    rows = []

    for card in cards:
        # 상태
        status = ""
        try:
            status = card.find_element(By.CSS_SELECTOR, STATUS_IN_CARD).text.strip()
        except Exception:
            for css in STATUS_FALLBACKS:
                try:
                    t = card.find_element(By.CSS_SELECTOR, css).text.strip()
                    if t:
                        status = t
                        break
                except Exception:
                    pass
        if not status:
            blob = (card.text or "")
            for kw in ["배송중", "배송완료", "상품준비중", "결제완료", "구매확정", "취소완료", "교환진행", "반품진행"]:
                if kw in blob:
                    status = kw
                    break

        # 이름(복수)
        names = []
        els = card.find_elements(By.CSS_SELECTOR, NAME_IN_CARD)
        if not els:
            for css in NAME_FALLBACKS:
                els = card.find_elements(By.CSS_SELECTOR, css)
                if els:
                    break
        for el in els:
            name = (el.text or el.get_attribute("title") or el.get_attribute("aria-label") or "").strip()
            if name:
                names.append(name)

        for nm in names:
            rows.append((nm, status))

    print(f"📦 수집 건수: {len(rows)}")
    for i, (n, s) in enumerate(rows, 1):
        print(f"{i}. {n} | {s}")

if __name__ == "__main__":
    if SCENARIO not in TOGGLES_BY_SCENARIO:
        print("SCENARIO 이름이 올바르지 않습니다:", SCENARIO)
        sys.exit(1)

    TOGGLES = TOGGLES_BY_SCENARIO[SCENARIO]
    PROFILE_TAG = SCENARIO  # 시나리오별 프로필을 분리해서 충돌/간섭 방지

    print(f"▶ 시나리오: {SCENARIO}")
    print(f"    - excludeSwitches/useAutomationExtension: {TOGGLES['EXCLUDE_SWITCHES']}")
    print(f"    - hide webdriver: {TOGGLES['HIDE_WEBDRIVER']}")
    print(f"    - UA/LANG flags: {TOGGLES['UA_LANG']}")
    print(f"    - use dedicated profile: {TOGGLES['USE_PROFILE']}")
    print(f"    - wait login marker: {TOGGLES['WAIT_LOGIN_MARKER']}")
    print(f"    - captcha detect: {TOGGLES['CAPTCHA_DETECT']}")

    driver = build_driver(TOGGLES, PROFILE_TAG)
    try:
        ok = ensure_session(driver)
        if ok:
            scrape_and_print(driver)
        else:
            print("❌ 세션 확보 실패(로그인/인증 필요).")
    finally:
        time.sleep(1)
        driver.quit()
# ===== 아래 블록을 파일 맨 아래 if __name__ == "__main__": 위/아래 편한 곳에 추가 =====

def run_once(scenario):
    global SCENARIO, TOGGLES, PROFILE_TAG
    SCENARIO = scenario
    TOGGLES = TOGGLES_BY_SCENARIO[SCENARIO]
    PROFILE_TAG = SCENARIO

    print("\n" + "="*72)
    print(f"▶ 시나리오 실행: {SCENARIO}")
    print(f"    - excludeSwitches/useAutomationExtension: {TOGGLES['EXCLUDE_SWITCHES']}")
    print(f"    - hide webdriver: {TOGGLES['HIDE_WEBDRIVER']}")
    print(f"    - UA/LANG flags: {TOGGLES['UA_LANG']}")
    print(f"    - use dedicated profile: {TOGGLES['USE_PROFILE']}")
    print(f"    - wait login marker: {TOGGLES['WAIT_LOGIN_MARKER']}")
    print(f"    - captcha detect: {TOGGLES['CAPTCHA_DETECT']}")

    driver = build_driver(TOGGLES, PROFILE_TAG)
    ok = False
    rows = []
    try:
        ok = ensure_session(driver)
        if ok:
            # 요소 대기
            wait = WebDriverWait(driver, ELEMENT_WAIT_SEC)
            try:
                wait.until(lambda d: d.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL))
            except Exception:
                pass
            cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
            # 간단히 건수만 확인(상세 출력은 기존 함수 사용 가능)
            print(f"   -> 카드 감지: {len(cards)}개")
            if cards:
                # 첫 카드만 샘플 출력
                n = min(1, len(cards))
                print("   -> 샘플 카드 텍스트:", cards[0].text[:120].replace("\n"," ") + "...")
            print("   -> 결과 요약 완료")
    finally:
        time.sleep(0.8)
        driver.quit()
    return ok

def run_ablation_suite():
    scenarios = [
        "baseline",
        "no_exclude_switches",
        "no_hide_webdriver",
        "no_ua_lang",
        "no_profile",
        "no_login_marker_wait",
        "no_captcha_detect",
    ]
    results = {}
    for sc in scenarios:
        ok = run_once(sc)
        results[sc] = "PASS" if ok else "FAIL"
        # 너무 빠른 연속 접근 방지
        time.sleep(1.2)

    print("\n=== 아블레이션 결과 요약 ===")
    for sc, res in results.items():
        print(f"{sc:24s} : {res}")

# 단발 실행 대신 전체 스위트 돌리려면 아래 한 줄 호출
run_ablation_suite()
```

```python
# -*- coding: utf-8 -*-
# 쿠팡 주문/배송 조회: 최소 구성 + 지정 셀렉터만 사용

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time, sys

# === 환경 설정 ===
CHROMEDRIVER = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"
ORDER_URL = "https://mc.coupang.com/ssr/desktop/order/list?requestYear=2024"

# 네가 지정한 정확한 셀렉터
ORDER_CARD_SEL = "div.sc-fimazj-0.gKYVxm"                         # 주문내역 제품 블럭
NAME_IN_CARD   = "span.sc-755zt3-1.sc-8q24ha-1.ksjYDU.ifMZxv"     # 제품명
STATUS_IN_CARD = "span.sc-755zt3-0.anmor"                         # 배송상태

# === 드라이버 설정 ===
service = Service(CHROMEDRIVER)
options = Options()
options.add_argument("--start-maximized")  # 창 최대화

driver = webdriver.Chrome(service=service, options=options)

# 핵심 우회: navigator.webdriver 숨김
driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
    "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined});"
})

# === 실행 ===
try:
    driver.get(ORDER_URL)

    # 주문 블럭 대기
    wait = WebDriverWait(driver, 10)
    try:
        wait.until(lambda d: d.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL))
    except Exception:
        pass
    time.sleep(0.5)

    cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
    if not cards:
        print("❌ 주문내역 블럭이 보이지 않습니다. 로그인 후 다시 실행하세요.")
        sys.exit(1)

    rows = []
    for card in cards:
        status = card.find_element(By.CSS_SELECTOR, STATUS_IN_CARD).text.strip()
        names = [el.text.strip() for el in card.find_elements(By.CSS_SELECTOR, NAME_IN_CARD) if el.text.strip()]
        for nm in names:
            rows.append((nm, status))

    # 출력
    print(f"📦 수집 건수: {len(rows)}")
    for i, (n, s) in enumerate(rows, 1):
        print(f"{i}. {n} | {s}")

finally:
    time.sleep(1)
    driver.quit()
```

```python
# -*- coding: utf-8 -*-
# 쿠팡 주문/배송: 최소 구성 + 쿠키 저장/재사용 + 지정 셀렉터만 사용

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time, os, pickle, json, sys

# ===== 환경설정 =====
CHROMEDRIVER = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"

ORDER_URL = "https://mc.coupang.com/ssr/desktop/order/list?requestYear=2024"
COOKIES_PKL = "coupang_cookies.pkl"   # 쿠키 저장 파일
COOKIES_JSON = "coupang_cookies.json" # (참고용) JSON 저장

# 네가 지정한 정확한 셀렉터
ORDER_CARD_SEL = "div.sc-fimazj-0.gKYVxm"                         # 주문내역 제품 블럭
NAME_IN_CARD   = "span.sc-755zt3-1.sc-8q24ha-1.ksjYDU.ifMZxv"     # 제품명
STATUS_IN_CARD = "span.sc-755zt3-0.anmor"                         # 배송상태

LOGIN_WAIT_SEC = 30   # 수동 로그인 대기(주문카드 등장까지)
ELEMENT_WAIT_SEC = 10 # 요소 대기

# ===== 드라이버 =====
service = Service(CHROMEDRIVER)
options = Options()
options.add_argument("--start-maximized")  # 너가 추가했던 옵션만 유지(심플)

driver = webdriver.Chrome(service=service, options=options)

# 핵심 우회: navigator.webdriver 숨김
driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
    "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined});"
})

# ===== 쿠키 유틸 =====
def save_cookies():
    cookies = driver.get_cookies()
    with open(COOKIES_PKL, "wb") as f:
        pickle.dump(cookies, f)
    with open(COOKIES_JSON, "w", encoding="utf-8") as f:
        json.dump(cookies, f, ensure_ascii=False, indent=2)
    print(f"✅ 쿠키 저장 완료 → {COOKIES_PKL}, {COOKIES_JSON}")

def load_cookies():
    if not os.path.exists(COOKIES_PKL):
        return False
    with open(COOKIES_PKL, "rb") as f:
        cookies = pickle.load(f)
    # 쿠키는 같은 도메인에서만 주입 가능 → 먼저 coupang.com으로 진입
    driver.get("https://www.coupang.com/")
    time.sleep(0.5)
    for c in cookies:
        try:
            driver.add_cookie(c)
        except Exception:
            pass
    print("🍪 쿠키 주입 완료")
    return True

def logged_in():
    try:
        els = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
        return any(e.is_displayed() for e in els)
    except Exception:
        return False

# ===== 메인 플로우 =====
try:
    # 1) 쿠키가 있으면 먼저 주입 후 주문내역으로
    used_cookie = load_cookies()
    driver.get(ORDER_URL)

    # 2) 로그인 확인(주문카드 등장)
    if not logged_in():
        if used_cookie:
            # 쿠키가 있었는데도 로그인 안 되었으면, 수동 로그인으로 폴백
            print("⚠️ 쿠키로는 로그인되지 않았습니다. 수동 로그인으로 전환합니다.")
        else:
            print("🔐 수동 로그인 필요")
        # 수동 로그인 대기(주문카드 등장까지)
        end = time.time() + LOGIN_WAIT_SEC
        while time.time() < end:
            if logged_in():
                print("✅ 로그인 성공 감지")
                save_cookies()  # 성공 시 쿠키 갱신 저장
                break
            time.sleep(1)
        if not logged_in():
            print("❌ 로그인 신호를 감지하지 못했습니다. 로그인 후 다시 실행하세요.")
            sys.exit(1)

    # ===== (선택) 스크랩 & 출력 =====
    wait = WebDriverWait(driver, ELEMENT_WAIT_SEC)
    try:
        wait.until(lambda d: d.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL))
    except Exception:
        pass
    time.sleep(0.3)

    cards = driver.find_elements(By.CSS_SELECTOR, ORDER_CARD_SEL)
    rows = []
    for card in cards:
        status = card.find_element(By.CSS_SELECTOR, STATUS_IN_CARD).text.strip()
        names = [el.text.strip() for el in card.find_elements(By.CSS_SELECTOR, NAME_IN_CARD) if el.text.strip()]
        for nm in names:
            rows.append((nm, status))

    print(f"📦 수집 건수: {len(rows)}")
    for i, (n, s) in enumerate(rows, 1):
        print(f"{i}. {n} | {s}")

finally:
    time.sleep(1)
    driver.quit()
```

```python
# -*- coding: utf-8 -*-
# Gmail: 전용 프로필로 로그인 유지 (최소 코드)

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import os, time

CHROMEDRIVER = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"
USER_DATA_DIR = r"C:\SeleniumProfiles\Gmail"  # 전용 프로필 폴더 (한 번 로그인하면 계속 유지)
GMAIL_URL     = "https://mail.google.com/mail/u/0/"

os.makedirs(USER_DATA_DIR, exist_ok=True)

options = Options()
options.add_argument("--start-maximized")
options.add_argument(f"--user-data-dir={USER_DATA_DIR}")
options.add_argument("--profile-directory=Profile_Auto")

driver = webdriver.Chrome(service=Service(CHROMEDRIVER), options=options)

# 핵심 1줄: 자동화 탐지 완화 (navigator.webdriver 숨김)
driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
    "source": "Object.defineProperty(navigator, 'webdriver', {get: () => undefined});"
})

driver.get(GMAIL_URL)

# 첫 실행만 화면에서 수동 로그인하세요(2FA 포함). 이후부터는 이 프로필로 자동 로그인됩니다.
# 필요하면 잠깐 대기만 두고 끝:
# time.sleep(60)

# driver.quit()  # 필요하면 종료
```
