```python
#GS샵(*혹은 본인이 자주 방문 및 구매하는 쇼핑몰) 로그인 후 주문배송조회 페이지의 상품명과 상태를 수집하세요.
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pyautogui
import pickle
import pymysql 

# db = pymysql.connect(
#     host = "localhost",
#     port = 3306,
#     user = "root",
#     passwd = "hithere4961",
#     db = "ecommerce_v2",
#     charset = "utf8"
# ) 


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("--start-maximized")
options.add_argument("--lang=ko_KR")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")


driver = webdriver.Chrome(service=service, options=options)

url = "https://www.musinsa.com/main/musinsa/ranking?gf=A&storeCode=musinsa&sectionId=200&contentsId=&categoryCode=000&ageBand=AGE_BAND_ALL"
driver.get(url)
time.sleep(2)
prdlinks = []
driver.execute_script("window.scrollTo(0, 3300)")
time.sleep(2)

links = driver.find_elements(By.CSS_SELECTOR, "a.sc-1t5ihy5-2.iFTPaE.gtm-select-item")
for link in links:
    # print(link.get_attribute("href"))
    prd_link=link.get_attribute("href") 
    prdlinks.append(prd_link)
    
for link in prdlinks : 
    url_2 = f"{link}"
    driver.get(url_2)
    time.sleep(3)  
    try :
        title = driver.find_element(By.CSS_SELECTOR, "span.text-title_18px_med.sc-1omefes-1.exqQRL.font-pretendard").text
        brand = driver.find_element(By.CSS_SELECTOR, "span.sc-12cqkwk-2.hgzZM span.text-body_14px_med.font-pretendard").text
        price = driver.find_element(By.CSS_SELECTOR, "span.text-title_18px_semi.sc-1hw5bl8-7.kXhdZT.text-black.font-pretendard").text
        discount_rate = driver.find_element(By.CSS_SELECTOR, "span.text-title_18px_semi.sc-1hw5bl8-6.hROMjI.text-red.font-pretendard").text
        review_count = driver.find_element(By.CSS_SELECTOR, "div.sc-hw7d9p-0.lecuxg.gtm-click-button span.text-body_13px_reg.underline.text-gray-600.font-pretendard").text
        print(title,brand,price,discount_rate,review_count)
    except : 
        pass 

time.sleep(2)
# driver.quit()
```
