```python
#인터파크 티켓에서 공연 카테고리를 클릭해 공연명과 일정 정보를 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
driver = webdriver.Chrome(service=service, options=options)

url = "https://www.facebook.com/profile.php?id=100079202510961&locale=ko_KR"
driver.get(url)
time.sleep(2)
elements = driver.find_elements(By.XPATH, "//div[contains(@data-pagelet, 'Timeline')]")
height = 500
for idx, element in enumerate(elements, start=1):
    driver.execute_script(f"window.scrollTo(0,{height})")
    time.sleep(2)
    title = element.find_element(By.XPATH, ".//div[@style='text-align: start;']").text
    height += 500
    print(title)
    if idx >= 5:
        break

time.sleep(1)

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import InvalidSessionIdException, WebDriverException
from webdriver_manager.chrome import ChromeDriverManager
import time

URL = "https://www.facebook.com/profile.php?id=100079202510961&locale=ko_KR"

# 전역 중복 방지 리스트
SEEN_POSTS = []

# 드라이버 생성
def new_driver():
    service = Service(ChromeDriverManager().install())
    options = Options()
    # options.add_argument("--headless=new")  # 필요하면 활성화
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--start-maximized")
    options.add_argument("--lang=ko_KR")
    options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                         "AppleWebKit/537.36 (KHTML, like Gecko) "
                         "Chrome/138.0.0.0 Safari/537.36")
    return webdriver.Chrome(service=service, options=options)

# 세션 보장 + 이동
def ensure_get(drv, url):
    try:
        drv.get(url)
        return drv
    except (InvalidSessionIdException, WebDriverException):
        try:
            drv.quit()
        except:
            pass
        drv = new_driver()
        drv.get(url)
        return drv

# 엘리먼트가 피드 뷰포트 내 보이는지 판정
def in_viewport(driver, el, container):
    return driver.execute_script("""
        const el = arguments[0], cont = arguments[1];
        if (!el || !cont) return false;
        const er = el.getBoundingClientRect();
        const cr = cont.getBoundingClientRect();
        const overlapX = Math.max(0, Math.min(er.right, cr.right) - Math.max(er.left, cr.left));
        const overlapY = Math.max(0, Math.min(er.bottom, cr.bottom) - Math.max(er.top, cr.top));
        const visibleArea = overlapX * overlapY;
        const elArea = Math.max(1, (er.width||0) * (er.height||0));
        return visibleArea / elArea > 0.2; // 20% 이상 보이면 보이는 것으로 간주
    """, el, container)

# 현재 화면(피드 컨테이너 뷰포트)에 보이는 텍스트만 수집
def crawl_visible_texts(driver, feed):
    out = []
    articles = feed.find_elements(By.XPATH, ".//div[contains(@role,'article')]")
    for a in articles:
        nodes = a.find_elements(By.XPATH, ".//div[@style='text-align: start;']")
        for n in nodes:
            if not in_viewport(driver, n, feed):
                continue
            txt = n.text.strip()
            if not txt:
                continue
            if txt not in SEEN_POSTS:
                SEEN_POSTS.append(txt)
                out.append(txt)
    return out

# -------- 실행 --------
driver = new_driver()
try:
    driver = ensure_get(driver, URL)

    wait = WebDriverWait(driver, 10)
    # 피드 컨테이너 (동적 대응: contains)
    feed = wait.until(EC.presence_of_element_located(
        (By.XPATH, "//div[contains(@data-pagelet,'Timeline')]")
    ))

    # 스크롤→수집 반복 (피드 한두 화면씩만)
    cycles = 6      # 반복 횟수
    chunk  = 1.2    # 피드 높이의 몇 배만큼 스크롤

    for _ in range(cycles):
        for t in crawl_visible_texts(driver, feed):
            print("-----")
            print(t[:500])
        # 피드 컨테이너 자체 스크롤
        driver.execute_script(
            "arguments[0].scrollTop = arguments[0].scrollTop + arguments[0].clientHeight * arguments[1];",
            feed, chunk
        )
        time.sleep(1.2)  # 로딩 대기

    # 마지막으로 한 번 더
    for t in crawl_visible_texts(driver, feed):
        print("-----")
        print(t[:500])

finally:
    driver.quit()

```
