#### 1. pandas 
- pandas 는 파이썬에서 사용할 수 있는 라이브러리 (VS 프레임워크)

- 왜 pandas인가?
- 1) 가장 오랜 기간 테이블 형태의 데이터를 다룰 때 사용되어왔다 => 검증 // 버그 거의 안정화됨
  2) 현업에서 가장 많이 사용한다
  3) 기능이 너무 압도적이다. => plotly 데이터 시각화 라이브러리와 호환성 => 웬만한 전처리 & 데이터 시각화
- pandas 를 통해서 실행하고자 하는 기본 기능 위주로 문법 기초 학습
- 학습한 결과를 토대로 => 2가지 실습 (*코로나 확진자 데이터를 전처리 & 시각화 실습 // 이커머스 쇼핑몰 데이터 전처리 & 시각화)
  
- raw data : raw => 원본 데이터 기준 전처리
- EDA : 탐색적 데이터 분석 => Exploratory Data analysis

#### 2.pandazs 를 학습할 때, 중요하게 생각해야하는 개념 
- series : 1차원 데이터 => 1개의 필드명 + 컬럼 한줄
- Dataframe : 2차원 데이터
- 0,1,2 => 인덱스 (*좌측 기준)
- 미,한,중 => 필드명/값 (*상단 기준)
- 2.1, 2.2 ,2.3 => 각각의 필드명에 속해있는 요소들의 집합 => "컬럼"
- 2.1,2.2,2.3,10,13,15 => values

- CRUD (create/read/update/delete)
- pandas 데이터 타입
  1) int64: 정수
  2) float64 : 실수
  3) bool : 논리형
  4) object: 객체-문자열/ 1개의 시리즈 안에 복수의 데이터 타입이 혼용되어 있을때
  5) datetime64: 날짜.시간 타입

#### 4. 탐색적 데이터 분석 과정을 통해 데이터 전처리 
- EDA : Exploratory Data Analysis
- 개념 및 이론 : 데이터 분석을 위해서 raw data를 다양한 각도에서 관찰하고, 해당 데이터를 이해하는 과정
   - 하지만, 데이터의 인사이트를 도출하거나 이해하는 정답은 존재하지 않는다는 것은 기본 전제
   - 그러나, 수많은 데이터 분석가들이 데이터를 읽으면서 내린 공통적인 결론 및 의견들을 조합해서 하나의 이론 등장 => EDA
     1. 데이터의 출처와 주제에 대한 이해
     2. 데이터의 크기 확인
     3. 데이터 구성요소(feature)의 속성(특징) 확인

- 가설 & 검증 => 가설 & 검증
- 상관계수 : 각 속성 간 상관관계를 확인할 수 있도록 만들어진 수학적 공식
- 피어슨 상관계수 : 1에 가까울 수록 상관관계가 높다
