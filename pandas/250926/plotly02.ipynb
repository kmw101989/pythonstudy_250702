```python
# 복합 라인 = Multiple Line 

import plotly.graph_objects as go 
import numpy as np 
import pandas as pd 

df = pd.DataFrame(
    {
        "A" : np.random.rand(50),
        "B" : np.random.rand(50)
    }
)
fig = go.Figure() 


fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df["A"],
        mode = "lines + markers + text" ,
        name = "A" ,
        showlegend = True ,
        text = df.index,
        textposition = "top center" 
    )
)
fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df["B"],
        mode = "lines + markers + text" ,
        name = "B",
        text = df.index,
        showlegend = True,
        textposition= "top center" 
    )
)
```
