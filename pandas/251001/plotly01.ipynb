```python
import pandas as pd 
import plotly.graph_objects as go 
import numpy as np 

df = pd.DataFrame(np.random.rand(10,2),columns=["A","B"])

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df.index, y=df["A"], name="A",text = df["A"],textposition="outside",texttemplate="%{y:.2f}"
    )
)
fig.add_trace(
    go.Bar(
        x=df.index, y=df["B"], name="B",text = df["B"] , textposition = "auto",texttemplate="%{y:.2f}"
    )
)
fig.update_layout(
    {
        "title":{
            "text":"Graph With <b>go.bar</b>",
            "x": 0.5,
            "y": 0.9,
            "font" : {
                "size" :20
            }
        },
        "xaxis":{
            "title" : "random number",
            "showticklabels" : True
        },
        "yaxis":{
            "title" : "A"
        },
        "width" : 800,
        "height" : 340
    }
)

fig.show()
```
