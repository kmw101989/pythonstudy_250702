```python
import os, re, pathlib, mimetypes
from urllib.parse import urlparse, unquote
import requests

URLS = [
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/PiHuRohsqRXE8RNp.%E1%84%91%E1%85%AD%E1%84%8C%E1%85%B5%20%2B%20Overview.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/7gahiYMQNGLunKhp.Background.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/RootvCBujHDRU8TD.Survey.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/9MKkqobZzbjrh9BR.Interview.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/dNUawESM9X9FeryJ.1st%20MVP.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/Bf5MrkdnUrS68Z4a.UT%20Insight.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/NXRvfyGAEQoev9ZX.Branding%20%2B%20Design%20System.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/BYJ73C3cEBsVFaP5.Service%20ui.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/EquG7nA8ySyLHjJk.home%20%28record%29%203.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/9iresJr4onQJQRWy.home%20%28record%29%204.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/AFAQeHEoBubvGy5d.folder.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/aVKWNeCBe57hnNXA.report.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/2CJKs7JviRAH9z96.goods%20%2B%20mockup.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/bSWZDtruPUF3mRmP.end1.png",
    "https://cdn-bastani.stunning.kr/prod/portfolios/30a66fa3-753f-41d4-a7af-c69a67a2fe62/contents/cWoaweeNcfZ6reRq.end%202.png",
]

SAVE_DIR = pathlib.Path("stunning_downloads")
SAVE_DIR.mkdir(exist_ok=True)

# 원본 페이지를 헤더에 넣으면 차단 확률↓ (없으면 빼도 됨)
REFERER = "https://notefolio.net/maygirl_gh5591/429754"

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36",
    "Accept": "*/*",
    "Referer": REFERER,
})

def sanitize(name: str):
    name = unquote(name)  # URL 인코딩 해제
    name = re.sub(r'[\\/:*?"<>|]+', "_", name)  # 윈도우 금지 문자
    name = name.strip().strip(".")              # 끝의 점 제거
    return name or "image"

def unique_path(base: pathlib.Path):
    if not base.exists():
        return base
    stem, suffix = base.stem, base.suffix
    i = 2
    while True:
        cand = base.with_name(f"{stem}_{i}{suffix}")
        if not cand.exists():
            return cand
        i += 1

def filename_from_response(resp, fallback: str):
    cd = resp.headers.get("Content-Disposition", "")
    if "filename=" in cd:
        # filename="xxx.png"
        m = re.search(r'filename\*=UTF-8\'\'([^;]+)', cd)
        if m:
            return sanitize(unquote(m.group(1)))
        m = re.search(r'filename="?([^";]+)"?', cd)
        if m:
            return sanitize(m.group(1))
    # URL에서 추출
    return sanitize(fallback)

def ext_from_content_type(resp):
    ctype = (resp.headers.get("Content-Type") or "").split(";")[0].strip()
    ext = mimetypes.guess_extension(ctype) or ""
    # .jpe 보정
    if ext == ".jpe":
        ext = ".jpg"
    return ext

ok, fail = 0, 0
for u in URLS:
    parsed = urlparse(u)
    basename = os.path.basename(parsed.path) or "image"
    basename = sanitize(basename)
    # 확장자 없으면 Content-Type으로 보완
    try:
        r = session.get(u, stream=True, timeout=30)
        r.raise_for_status()
        # 확장자 없는 경우 보완
        root, ext = os.path.splitext(basename)
        if not ext:
            guessed = ext_from_content_type(r) or ".png"
            basename = root + guessed
        out = unique_path(SAVE_DIR / basename)
        with open(out, "wb") as f:
            for chunk in r.iter_content(chunk_size=1 << 14):
                if chunk:
                    f.write(chunk)
        ok += 1
        print(f"[OK] {u} -> {out}")
    except Exception as e:
        fail += 1
        print(f"[FAIL] {u} ({e})")

print(f"\n완료: 성공 {ok} / 실패 {fail}. 저장 폴더: {SAVE_DIR.resolve()}")
```
