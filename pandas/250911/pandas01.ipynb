#### pandas 라이브러리 : 전처리 & 기본그래프 시각화 최적화 
- 파이썬에서 문서 처리하는 기본.기초 지식 학습
- 특정 파일 오픈 -> 파일 읽기 || 파일 쓰기 -> 파일 닫기 

- 파일 처리 내장함수 -> open(실행하고자 하는 경로, 처리 모드 옵션, 인코딩 옵션) 
- 파일 처리 모드 옵션 ->: r(ead) / w(rite) : 쓰기 -> 기존 파일에 어떤 값을 쓰기위한 목적 / a(dd) -> 새로운 데이터 추가 
- 인코딩 옵션 -> 웹을 통해서 주고 받은 문서 파일 -> 영어 -> 한글 + 특수문자 + 이모지 
- 우리의 언어로 작성한 문자열 -> 인코딩 -> 컴퓨터 
- utf-8-sig

- 파일 포맷 : 데이터 분약 가장 빈번하게 많이 사용하는 포맷 
- csv= 엑셀 = 행 & 열 ,json = 프론트 & 백엔드 ,xml = 크롤링 , api 
- csv = comma-separated values = 실제 데이터 취급 시, 정말 많이 볼 수 있는 데이터 포맷 양식 
- xml = Extensible Markup Lang = 유연한 마크업 언어로 되어있는 데이터 포맷 양식 
- json = Javascript Object Notation = 클라이언트와 서버가 데이터를 주고 받을 때 사용하는 공식 데이터 포맷
