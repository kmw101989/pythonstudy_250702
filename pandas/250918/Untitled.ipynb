```python
import plotly.express as px 
import numpy as np 
import pandas as pd 

df = px.data.gapminder()

fig = px.scatter(
    df,
    x="gdpPercap", 
    y="lifeExp",
    size="pop", 
    color="country",
    hover_name="continent", 
    animation_frame="year", 
    animation_group="country",
    log_x=True, size_max=60
)

fig.show()
```
