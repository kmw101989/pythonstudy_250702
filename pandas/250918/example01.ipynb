```python
# 함수 선언문 시작 
def add(a,b) : 
    return a+b 

# 함수 호출 
print(add(1,2))
```

```python
# 함수 => 람다 식 => 익명함수 
# 함수의 이름을 없애는 대신 해당 함수를 어떠한 변수 안에 담아야 함 

add = lambda a,b: a+b
add(1,2)
```
