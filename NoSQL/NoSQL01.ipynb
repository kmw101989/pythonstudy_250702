#### 2. NoSQL 이해 
- NoSQL은 스키마 & Join 등의 문법 및 관리시스템이 존재하지 않음
- 현재 작업하는 업무가 95% read , 5% write 라면 : RDBMS 써라
- 현재 작업하는 업무에서 50%이상 write 라면 : NoSQL + SQL

#### 9. SQL JOIN VS NoSQL $lookup
1) SQL => RDBMS (*관계형 데이터베이스 관리 시스템)
   - schema 개념 // 테이블을 생성, 필드릉 선언 => Primary key, Foreign key => index
   - INNER JOIN : 서로 다른 테이블에서 교집합 (*매칭되는 행만 가져옴)
   - LEFT OUTER JOIN : 메인이 되는 왼쪽 테이블을 모두 가져오고, 서브가 되는 오른쪽 테이블은 매칭 되는 값만 가져옴
   - RIGHT OUTER JOIN : LEFT OUTER JOIN 의 반대개념

2) NoSQL => 기본적으로 무조건 LEFT OUTER JOIN 의 형태를 가지고 있다.

3) 
```
db.comments.aggregate([
    {
        $lookup:
            {
                from :"movies" // 조인할 대상의 컬렉션 이름
                localField: "movie_id"// 현재 컬렉션에서 조인에 사용할 필드
                foreignField: "_id" // 조인할 컬렉션에서 조인에 사용할 필드
                as: "movie"// 조인된 결과를 저장할 필드 이름 
            }
    }
])

#### $$ 가 들어간 사용자 정의 변수 = 시스템 변수 
- \$$ROOT : 현제 루트 폴더 문서 전체를 찾아오고자 할 때
- \$$NOW : 현재 날짜, 시간 조회
- \$$THIS : 배열 순회 중 현재 요소
