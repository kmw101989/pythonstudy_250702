```python
from bs4 import BeautifulSoup as bs
import requests
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

for page_num in range(8):
    if page_num == 0 : 
        res = requests.get("https://davelee-fun.github.io/")
    else :
        res = requests.get(f"https://davelee-fun.github.io/page{page_num+1}/")
    soup  = bs(res.content, "html.parser")
    data = soup.select("div.card-body")
    for item in data : 
        category = item.select_one("h2.card-title").text.strip().replace("관련 상품 추천", "")
        product = item.select_one("h4.card-text").text.replace("상품명:","").strip()
        print(category,product)
        data = dict()
        data["category"] = category
        data["product"] = product  
        collection.insert_one(data)
        
client.close() 
```

```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

documents = list(collection.find({}))
for document in documents : 
    print(document)
```

```python
#category 필드에 있는 모든 카테고리 종류 조회 
daveshop = db.daveshop

list(daveshop.distinct("category"))
```

```python
#모든 상품 (product) 조회 
products = daveshop.find({},{"product":1,"_id":0})
for product in products:
    print(product["product"])
```

```python
# 매트리스 커버 카테고리의 모든 상품 조회 
products = daveshop.find({"category":"매트리스커버"},{"_id":0,"product":1})
for product in products:
    print(product["product"])
```

```python
pipeline = [
    {"$match":{"category":"매트리스커버"}},
   {"$project": {"_id" : 0, "product":1}}
    
]
products = daveshop.aggregate(pipeline)

for product in products : 
    print(product)
```

```python
# 각 카테고리 별 상품 수 

pipeline = [
    {"$group" :{
        "_id" : "$category",
        "sumProduct" : {"$sum":1}
    }}
]

datas = daveshop.aggregate(pipeline)
for data in datas : 
    print(data)
```

```python
#상품명에 "핑크" 가 포함된 상품 조회
products = collection.find({"product":{"$regex":".*핑크*."}} ,{"_id":0,"product":1})

for product in products:
    print(product["product"])
```
