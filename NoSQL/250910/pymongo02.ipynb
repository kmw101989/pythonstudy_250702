```python
from pymongo import MongoClient 

client = MongoClient("mongodb://localhost:27017")

db = client.sample_mflix
movies = db.movies

print(movies)
```

```python
# find() => aggregation Framework 문법을 사용하면 pipeline 생성해서 복잡한 데이터처리를 순차적으로 처리할 수 있다. 

# 1. $match => 특정조건에 해당되는 문서를 찾아오고자 할 때 
# 2. $limit => find().limit(2) 동일한 기능 

for movie in movies.aggregate([{"$match":{"genres":"Action"}},{"$limit": 1}]) : 
    print(movie) 
```

```python
# $group : SQL 에서의 GROUP BY 와 동일한 기능 

pipeline = [{"$group": {"_id":"$directors", "count":{"$sum":1}}}]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
pipeline = [
    {"$sort": {"title":1}}, {"$limit":3}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
pipeline = [
    {"$project":{"_id":0, "title":1,"genres":1}}, {"$limit":3}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
pipeline =[
    {"$unwind": "$genres"},{"$limit":3}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
pipeline = [
    {"$group":{"_id":"$directors" , "count":{"$sum":1}}}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
pipeline = [
    {"$group":{"_id": "$directors","average_rating":{"$avg":"$imdb.rating"}}}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
#collection에 있는 영화의 개수 계산 

count = movies.count_documents({})

print(count)
```

```python
# 각 장르에 대한 영화 갯수 
pipeline = [
    {"$unwind":"$genres"},
    {"$group":{"_id":"$genres","genreCounts":{"$sum" :1}}}
]
genre_counts = list(
    movies.aggregate(pipeline)
)

for genre in genre_counts : 
    print(f"장르 : {genre['_id']}, 영화 수: {genre['genreCounts']}")
```

```python
# 2015 년 이후에 개봉한 영화를 제목으로 정렬하여 나열 

pipeline = [
    {"$match":{"year":{"$gte":2015}}},
    {"$sort": {"title":1}},
    {"$project":{
        "_id":0,
        "title":1,
        "year":1
    }}
]

for movie in movies.aggregate(pipeline):
    print(f"제목: {movie["title"]},  {movie["year"]}년")
```

```python
# 가장 많은 영화를 제작하는 상위 5개 국가 찾기 
pipeline = [
    {"$unwind":"$countries"},
    {"$group":{"_id":"$countries","count":{"$sum":1}}},
    {"$sort":{"count":-1}},{"$limit":5}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
# 2000년 이후 영화의 연도별 평군 imdb 평점 찾기 
pipeline = [
    {"$match":{"year":{"$gte":2000}}},
    {"$group":{"_id":"$year", "avgRating":{"$avg":"$imdb.rating"}}},
    {"$sort":{"_id":1}}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```
