```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from openpyxl import Workbook
import time

# ✅ 크롬 드라이버 설정
service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--disable-features=PasswordBreachDetection")
options.add_argument("--disable-popup-blocking")

driver = webdriver.Chrome(service=service, options=options)
driver.get("https://eyejokerdb.com/benchmarks/keyboard")
time.sleep(3)

viewport_height = driver.execute_script("return window.innerHeight;")

# ✅ 엑셀 생성
wb = Workbook()
ws = wb.active
ws.title = "Keyboard Benchmarks"

columns = [
    "종합 등급", "제품명", "눈쟁이 추천 등급", "구매", "가격",
    "래피드 트리거", "종합 응답속도", "실성능", "동시입력 딜레이",
    "유효 폴링레이트", "최저 입력지점", "하단 데드존"
]
ws.append(columns)

seen = set()

def parse_block(block_text):
    lines = block_text.strip().split("\n")
    if len(lines) < 2:
        return None
    # ✅ 티저 제외 (추천 문구 없는 경우)
    if not any(tag in block_text for tag in ["추천", "비추천"]):
        return None



# ✅ 초기 + 스크롤 반복
for i in range(10):  # 스크롤 10번
    elements = driver.find_elements(By.XPATH, "/html/body/div[2]/div[6]/div/div[1]/div[3]/div/div[2]/div/div")
    
    for el in elements:
        full_text = el.text.strip()
        # ✅ 한 블록에 여러 제품이 붙어 있을 경우 처리
        blocks = full_text.split("\n\n")
        for block in blocks:
            if block not in seen:
                seen.add(block)
                parsed = parse_block(block)
                if parsed:
                    ws.append([parsed[col] for col in columns])

    # ✅ 스크롤
    driver.execute_script(f"window.scrollBy(0, {viewport_height});")
    time.sleep(1)

driver.quit()

# ✅ 엑셀 저장
output_file = "keyboard_benchmark_final.xlsx"
wb.save(output_file)
print(f"✅ 크롤링 완료 → {output_file} 저장됨!")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pyautogui
from openpyxl import Workbook

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--disable-features=PasswordBreachDetection")
options.add_argument("--disable-popup-blocking")

driver = webdriver.Chrome(service=service, options=options)

url = "https://eyejokerdb.com/benchmarks/keyboard"
driver.get(url)

time.sleep(2)
pyautogui.click(x=1200 , y=470)

time.sleep(2)
# ✅ 뷰포트 높이 가져오기
viewport_height = driver.execute_script("return window.innerHeight;")
print(f"뷰포트 높이: {viewport_height}px")

# ✅ 엑셀 워크북 생성
wb = Workbook()
ws = wb.active
ws.title = "Keyboard Data"
ws.append(["크롤링 텍스트"])  # 첫 번째 열 헤더

all_text = []  # 전체 텍스트 저장

for i in range(10):
    elements = driver.find_elements(By.XPATH, "/html/body/div[2]/div[6]/div/div[1]/div[3]/div/div[2]/div/div")
    for element in elements:
        text = element.text.strip()
        if text and text not in all_text:  # 중복 제거
            all_text.append(text)
            ws.append([text])  # 엑셀에 추가
    driver.execute_script(f"window.scrollBy(0, {viewport_height});")
    time.sleep(1)

# ✅ 엑셀 저장
output_file = "keyboard_data.xlsx"
wb.save(output_file)
print(f"✅ 크롤링 완료 → {output_file} 저장됨!")

time.sleep(2)
driver.quit()
```
