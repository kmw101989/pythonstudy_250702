#### 다음 웹 사이트에 접속해서 selenium을 활용해 크롤링해오세요. 
- https://crawlingtest.netlify.app/
- id = dev_course_list 의 자식 class = course 만 출력 

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("--remote-debugging-port=9222")  # DevToolsActivePort 문제 해결
options.add_argument("lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

url = "https://crawlingtest.netlify.app/"
driver.get(url)

elements = driver.find_elements(By.CSS_SELECTOR, "#dev_course_list .course a")
for element in elements:
    print(element.text.strip()) 
time.sleep(1)

driver.quit()
```
