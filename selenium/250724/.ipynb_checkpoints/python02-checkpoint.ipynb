#### HTML 크롤링을 위한 XPATH 기본 형태 
- //tagname[@Attribute="value"]
- /가 하나라면 절대경로 : 예) /html/body/div/a
- // 문서 내에서 바로 검색 하라는 의미 예) //h1 문서에서 바로 h1 태그를 찾아와라
- //*[@href] : 문서상에서 href를 모두 찾음 
- //a[@href="https://www.google.com"] : 문서 내 a 태그를 찾아오는데, href 속성값으로 구글 닷컴을 찾아와라
- (//a)[3] : 문서 내 a 태그를 찾아오는데 세번째 a만 (0번부터 세지 않음, 1부터)
- (//table)[last()] : 문서 내 table 태그 중 마지막번째만 찾아옴
- (//a)[position() < 3] : 문서 내 a 태그를 찾아오는데, 인덱스가 < 3 인것들만 찾아옴
- //div[@*] : 문서 내 div태그를 찾아오는데, 어떤 속성이든 상관없이 속성이 있으면 찾아옴

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
from urllib.request import urlretrieve as ut 
import pyautogui
from openpyxl import Workbook


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=service, options=options)

url = "https://crawlingtest.netlify.app/"
driver.get(url)

elements = driver.find_elements(By.XPATH,  "//td")
rowList = []
for num in range(0,3): 
    # print(element.text)
    listdata = (elements[3*num].text, elements[3*num+1].text , elements[3*num+2].text)
    
    rowList.append(listdata)


time.sleep(2)
#------------종료---------

driver.quit()


wb = Workbook()
ws = wb.active 
ws.title = "크롤링 테스트"
ws.column_dimensions['A'].width = 15
ws.column_dimensions['B'].width = 70
ws.column_dimensions['C'].width = 30

for row in rowList : 
    ws.append(row)

wb.save("크롤링테스트.xlsx")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pyautogui

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--disable-features=PasswordBreachDetection")
options.add_argument("--disable-popup-blocking")

driver = webdriver.Chrome(service=service, options=options)

url = "https://divjason.github.io/sellenium-test/"
driver.get(url)

#----------로그인 정보 입력------------
element = driver.find_element(By.XPATH, "//input[@id='username']")
element.clear()
element.send_keys("error@error.com")

element= driver.find_element(By.XPATH, "//input[@id='password']")
element.clear()
element.send_keys("1234")
time.sleep(1)

#----------로그인 버튼 클릭------------
element = driver.find_element(By.XPATH, "//input[@type='submit']")
element.click()
time.sleep(1)


#----------알림 버튼 강제 클릭-----------
pyautogui.click(x=1140 , y=335)
time.sleep(1)


#----------로그인 후 메세지 출력---------
element = driver.find_element(By.XPATH,"//div[@class='message']")
print(element.text)


time.sleep(1)
#------------종료---------
driver.quit()
```
