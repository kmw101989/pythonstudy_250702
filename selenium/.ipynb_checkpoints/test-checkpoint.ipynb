```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import pandas as pd
import time
import pyautogui

# ✅ ChromeDriver 설정
service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=service, options=options)

# ✅ URL
url = "https://m.bunjang.co.kr/shop/3541801/reviews?"
driver.get(url)

# ✅ 스크롤 & 알림 닫기
for i in range(3):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(0.1)
    pyautogui.click(x=930, y=470)
    time.sleep(0.1)

time.sleep(2)

# ✅ 리뷰 개별 블록
review_blocks = driver.find_elements(By.XPATH, "//*[@id='root']/div/div/div[4]/div/div[5]/div/div[2]/div")

data = []

for block in review_blocks:
    lines = block.text.strip().split("\n")
    if len(lines) < 3:
        continue

    nickname = lines[0].strip() if len(lines) > 0 else "None"
    date = lines[1].strip() if len(lines) > 1 else "None"
    product = lines[2].strip() if len(lines) > 2 else "None"

    # 리뷰 텍스트 찾기
    review_text = "None"
    keywords = []
    for line in lines[3:]:
        if len(line) >= 2 and "상품" not in line and "배송" not in line and "+" not in line:
            review_text = line
        else:
            keywords.append(line)

    keywords = [k for k in keywords if k and "+" not in k]  # "+2" 같은 값 제거
    keyword_text = ", ".join(keywords) if keywords else "None"

    data.append({
        "닉네임": nickname,
        "날짜": date,
        "상품명": product,
        "리뷰내용": review_text,
        "키워드": keyword_text
    })

driver.quit()

# ✅ 엑셀 저장
df = pd.DataFrame(data)
df.to_excel("bunyang_reviews_clean.xlsx", index=False)
print("✅ 저장 완료: bunyang_reviews_clean.xlsx")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import pandas as pd
import time
import pyautogui

# ✅ ChromeDriver 설정
service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=service, options=options)

# ✅ URL
url = "https://m.bunjang.co.kr/shop/3541801/reviews?"
driver.get(url)

# ✅ 스크롤 & 알림 닫기 반복
for i in range(5):  # 필요시 횟수 늘려주세요
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(0.3)
    pyautogui.click(x=930, y=470)  # 버튼 좌표
    time.sleep(0.3)

time.sleep(2)

# ✅ 개별 리뷰 블록
review_blocks = driver.find_elements(By.XPATH, "//*[@id='root']/div/div/div[4]/div/div[5]/div/div[2]/div")

data = []
keyword_patterns = ["상품", "배송", "번개톡", "포장"]

for block in review_blocks:
    lines = block.text.strip().split("\n")
    if len(lines) < 3:
        continue

    nickname = lines[0].strip()
    date = lines[1].strip()
    product = lines[2].strip()

    # ✅ 리뷰 텍스트는 상품명 바로 아래 줄만 확인
    review_text = "None"
    if len(lines) > 3:
        candidate = lines[3].strip()
        if not any(k in candidate for k in keyword_patterns) and "+" not in candidate:
            review_text = candidate

    # ✅ 키워드 수집 (4번째 줄 이후)
    keywords = []
    for line in lines[3:]:
        if any(k in line for k in keyword_patterns):
            keywords.append(line)
    keyword_text = ", ".join(keywords) if keywords else "None"

    data.append({
        "닉네임": nickname,
        "날짜": date,
        "상품명": product,
        "리뷰내용": review_text,
        "키워드": keyword_text
    })

driver.quit()

# ✅ 엑셀 저장
df = pd.DataFrame(data)
df.to_excel("bunyang_reviews_fixed.xlsx", index=False)
print("✅ 크롤링 완료 → bunjang_reviews.xlsx 저장됨!")
```
