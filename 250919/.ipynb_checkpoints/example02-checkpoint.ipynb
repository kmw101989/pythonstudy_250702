```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=service, options=options)
results = []
time.sleep(2)
for i in range(2):
    if i == 0 : 
        url = "https://books.toscrape.com/catalogue/category/books/mystery_3/index.html"
    else:
        url = f"https://books.toscrape.com/catalogue/category/books/mystery_3/page-{i+1}.html"
    driver.get(url)
    category = driver.find_element(By.CSS_SELECTOR,"div>h1").text
    elements = driver.find_elements(By.CSS_SELECTOR,"li.col-xs-6.col-sm-4.col-md-3.col-lg-3")
    for element in elements :
        title = element.find_element(By.CSS_SELECTOR, "a>img").get_attribute("alt")
        price_gbp = float(element.find_element(By.CSS_SELECTOR,"p.price_color").text.replace("£",""))
        stock = element.find_element(By.CSS_SELECTOR,"p.instock.availability").text
        rating = element.find_element(By.CSS_SELECTOR,"p.star-rating").get_attribute("class").replace("star-rating","").strip("")
        product_url = element.find_element(By.CSS_SELECTOR , "h3>a").get_attribute("href")
        
        
        result = {
            "category": category,
            "title": title,
            "price_gbp": price_gbp,
            "stock": stock,
            "rating": rating,
            "product_url": product_url
        }
        results.append(result)
for result in results :
    print(result)
    
time.sleep(2)
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=service, options=options)

results = []
links = [] 
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

def rating_num (rating) : 
    if not rating: 
        return None
    return rating_map.get(rating,0)


time.sleep(2)
url = "https://books.toscrape.com/catalogue/category/books_1/index.html"

driver.get(url)
time.sleep(2)

cats = driver.find_elements(By.CSS_SELECTOR, "ul.nav.nav-list>li>ul>li>a")
for cat in cats: 
    link = cat.get_attribute("href").replace("index.html","")
    links.append(link)
time.sleep(2)

for link in links : 
    for i in range(10) :
        if i == 0 :
            urls = f"{link}index.html"
        else : 
            urls = f"{link}page-{i+1}.html"
        try:
            driver.get(urls)
            category = driver.find_element(By.CSS_SELECTOR,"div>h1").text
            elements = driver.find_elements(By.CSS_SELECTOR,"li.col-xs-6.col-sm-4.col-md-3.col-lg-3")
            for element in elements :
                title = element.find_element(By.CSS_SELECTOR, "a>img").get_attribute("alt")
                price_gbp = float(element.find_element(By.CSS_SELECTOR,"p.price_color").text.replace("£",""))
                stock = element.find_element(By.CSS_SELECTOR,"p.instock.availability").text
                rating = element.find_element(By.CSS_SELECTOR,"p.star-rating").get_attribute("class").replace("star-rating","").strip()
                product_url = element.find_element(By.CSS_SELECTOR , "h3>a").get_attribute("href")
                
                
                result = {
                    "category": category,
                    "title": title,
                    "price_gbp": price_gbp,
                    "stock": stock,
                    "rating": rating_num(rating),
                    "product_url": product_url
                }
                results.append(result)
        except:
            break
for result in results :
    print(result)
                
driver.quit()
```
