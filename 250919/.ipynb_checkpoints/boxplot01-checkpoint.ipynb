3) 전체 데이터를 구성하고 있는 가각의 요소들을 어떻게 지칭할 것인가
    - 최소값 - MIN
    - 1사분위수 - Q1
    - 중앙값 - Median
    - 3사분위수 - Q3 
    - 최대값 - MAX
    - IQR (*interquartile Range) -> 1사분위 ~ 3사분위 사이의 값
    - 이상치
       - 상한 이상치 (Upper Fence) 
       - 하한 이상치 (Lower Fence)
    -  

data = [3,5,7,8,12,13,14,18,20,21]
Min = 3 
MAX = 21 
Median : 12.5 
Q1 : 6.5
Q3 : 18.5
IQR = 18.5 - 6.5 = 12 
Lower Fence : 6.5 - (1.5 * 12) = -11.5 
Upper Fence 36.5
