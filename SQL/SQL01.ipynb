### SQL 학습 전, 반드시 알고가야하는 개념정의
1. 데이터 (Data)
2. 데이터베이스
3. 스키마 (Schema)
4. 테이블 (Table)
5. SQL
    

#### 1. 데이터 (Data) 
- 수집 가능한 값
- 숫자 : 23, 5.9
- 문자 : "david", "2025-07-31"
- 상태 : True , False
- 데이터 그 자체로는 의미가 부족함 -> 구조화가 되어야 데이터는 비로소 가치를 갖게 됨


#### 2. 데이터베이스 (Database = DB)
- 여러 가지 다양한 데이터를 구조적이고 체계적으로 저장 및 관리할 수 있도록 해주는 저장소
- 종류 
   - 관계형 데이터베이스 (RDBMS -> Oracle // MySQL)
   - 비관계형 데이터베이스 (DMBS -> MongoDB)
- 오프라인 가게를 운영하는데 일일 고객 주문 정보, 특정 기간동안 사용한 재고 수량 등

#### 3. 스키마 (Schema)
- 데이터베이스의 구조를 정의한 것
- 데이터베이스를 어떻게 구조화할 것인지를 정의해놓은 일종의 설계도
- 스키마는 데이터베이스 내부에 존재하는 논리적 구조 

#### 4. 테이블 (Table)
- 관계형 데이터베이스 안에 저장된 실제 데이터의 저장 단위
- 구성요소
    - 열 (Column) = 속성 (ex. 고객의 나이, 성별 ,이메일 등)
    - 행 (Row) = 레코드 (ex. 한 명의 고객)


#### 5. SQL (Structured Query Language)
- 관계형 데이터베이스를 정의, 조작,제어 할 수 있도록 도와주는 표준 언어

종류 
 - DDL (Data Definitin Language) : Create , drop ,alter
 - DML (Data Manipulation Lang) : SELECT, INSERT , UPADATE , DELETE
 - DCL (Data Control lang) : Grant, REVOKE

데이터: 가장 하위 개념 . 원초적인 정보

### SQL 중급 단계 
1) FOREIGN KEY (외래 키) <-> PRIMARY KEY (고유 키)

2) HAVING
   - HAVING 절을 활용해서 조건 비교 !!
   - WHERE 절을 활용해서도 가능한데 , 굳이 왜 필요할까?
   - 특정 테이블 내 데이터를 그룹핑하는 GROUP BY를 적용한 상태에서 집계함수가 포함된 조건식을 사용하려고 할 때, 반드시 HAVING 절을 써야 함

3) JOIN
   - 서로 다른 테이블을 하나의 연결 테이블로 활용하고자 할 때 JOIN
   - 복수의 테이블을 관리하는 경우가 빈번 : 각각 테이블 간 연결고리를 FOREIGN KEY 를 만들어놓고 , 해당 테이블간 연결하는 경우가 숙명적으로 많을수밖에 없음 
   - inner join : 서로 다른 테이블간 매칭되는 데이터에 한해서 해당 데이터를 기준으로 연결
   - 서로 다른 테이블을 하나로 연결하기 위해서는 어떤 데이터를 기준으로 연결할 것인가 ? 

   - outer join : 사로 매칭되지 않는 데이터도 모두 찾아옴 !!
        - left OUTER JOIN : 먼저 언급된 (*LEFT data = 부모데이터) 를 기준으로 다른 테이블을 연결 , 자식데이터에 부모데이터와 매칭되지 않은 데이터를 누락시키지 않고                             전부 찾아옴 

        - right OUTER JOIN : 자식데이터를 기준으로 다른 테이블을 연결 , 마찬가지로 매칭되지 않은 데이터들을 누락시키지 않고 전부 찾아옴

4) SubQuery
   - SQL 문 안에 포함되어있는 SQL 문 => 중첩 SQL 문

5) INDEX
    - 파이썬 > 리스트 > 리스트 자료구조 내부에 있는 아이템들에게 각각 인덱스값을 부여
    - SQL 구문을 활용해서 컬럼 안에 입력되어있는 데이터들에게 개별적인 인덱스값을 부여 -> 값을 서칭해서 찾아오는 빈도 잦음 -> 너무 많은 시간,과정
    - 인덱스 종류
        - 클러스터형 인덱스 : 영어 사전과 같은 형태로 데이터를 정렬한다고 생각하면 됨 ex) PRIMARY KEY // FOREIGN KEY // UNIQUE 컬럼 
        - 보조 인덱스 : 데이터는 그대로 둔 상태에서 마치 책으로 비유하면 책 끄툽분에 있는 찾아보기 색인 페이지 정도로 생각하면 됨

### MySQL 중.고급 문법 및 단계 

- 다양한 SQL 함수
   - 문자열 함수 :
     1) 문자의 길이를 반환하는 함수 -> LENGTH(string)
     2) UPPER(string) : 특정 문자를 대문자로 변환시키는 함수
     3) LOWER(string) : 특정 문자를 소문자로 변환시키는 함수
     4) CONCAT(string1, string2 ....) : 복수의 문자열을 하나로 연결하는 함수
     5) SUBSTRING(string,start,length) : 문자열에서 일부 문자열만 추출해주는 함수
        
   - 날짜/시간 함수:
     1) NOW() : 현재 날짜와 시간을 반환하는 함수
     2) CURDATE() : 현재 날짜를 반환하는 함수
     3) CURTIME() : 현재 시간을 반환하는 함수
     4) DATE_ADD(date, INTERVAL unit) : 특정 날짜에서 간격을 추가해주는 함수
        - MONTH / YEAR / DAY / HOUR / SECOND / MINUTE
     5) DATE_SUB(date, INTERVAL unit) : 특정 날짜에서 간격을 빼주는 함수
     6) EXTRACT(추출해오고자 하는 값 = field FROM 추출대상= source)
     7) YEAR(), MONTH() , DAY(), HOUR(), MINUTE(), SECOND()
     8) DAYOFWEEK() : 날짜의 요일값 반환 - 숫자형태 : 1->일요일 ... 7->토요일
     9) DATE_FORMAT() : 형식지정자로 값을 받아서 요일을 출력할 수 있음 -> %W %a
        - %Y : 네 자리 연도 -2025
        - %y : 연도의 마지막 두자리 -25
        - %M : 영문 월 June
        - %m : 월을 두 자리 수로 표기 : 01,02..
        - %c : 월을 한 자리 숫자로 표기 : 1,2,3,4,5..
        - %D : 일을 숫자 + 영문 접미사 1st, 2nd,3rd,4th..
        - %d : 일을 두 자리 숫자 - 31,30,29..
        - %H : 시간을 24시간 형식의 두 자리로 표기 15 18 20
        - %h : 시간을 12시간 형식의 두 자리로 표기 03 06 08
        - %i : 분을 두 자리 숫자로 표기
        - %s : 초를 두 자리 숫자로 표기 
     11) TIMESTAMPDIFF(unit) : 두 날짜 혹은 시간 값의 차이를 계산 할 때 사용하는 함수
     12) ABS(number) : Absolute 절대적인 -> 절대값 추출
     13) CEIL(number): 올림처리
     14) FLOOR(number): 버림처리
     15) ROUND(number) : 반올림 -> ROUND(숫자A, 숫자B) -> 숫자 B가 최종적인 반올림의 결과값이 되도록 함
     16) SQRT(number) : 해당 숫자의 제곱근을 반환하는 함수 

- SubQuery 고급 활용 : SQL 쿼리 문 안에 또 다른 SQL 쿼리문이 존재하는 것
    - 기본적으로 서브쿼리는 최근 문법인 JOIN ON 이라는 문법으로 대체 가능
    - 하지만, 서브쿼리를 여전히 메인 문법으로 사용하고 있는 조직 존재
    - 과거 문법이어도 충분히 인지해야 하는 경우가 빈번하게 발생하기 때문에 꼭 학습
      1) 서브쿼리는 특정 테이블의 필드값을 어떤 조건 하에 다른 테이블과 연결하여 값을 찾아와야하는 경우에 굉장히 많이 사용됨
      2) 서브쿼리를 FROM 절에서 사용할 때에는 반드시 AS 구문을 사용해야함
      3) 상관 서브쿼리 : 기존 서브쿼리와 달리, 데이터를 참고하고자 하는 테이블이 동일한 경우, 해당 테이을 사용하는 위치에 따라 상위에서 관리되고 있는 테이블을 참조할                          수도 있는데,이 상황을 상관 서브쿼리라고 부름 
- 집합 연산자
    - 합집합 : 서로 다른 집합을 하나로 합려서 사용할 때 -> 서로 다른 테이블을 하나로 합쳐서 사용할 때 - > 서로 다른 테이블을 하나로 합쳐서 사용할 때

        2) UNION ALL : 중복행 까지 포함
    - 교집합 : 사로 다른 집합의 공통 집합을 추출 할 때
        1) INTERSECTION : 서로 다른 테이블의 공통 부분만 추출할 때
    - 차집합 : 서로 다른 테이블을 놓고 결합할 때, 어느 한쪽 영역에만 속해있는 부분만 가져올 때
        - EXCEPT : 서로 다른 테이블 결합 시, 어느 한쪽 데이터만 찾아오고 싶을 때  
  
- Transaction / COMMIT / ROLLBACK
    - 이미 만들어져있는 DB > TABLE 존재 -> TABLE안에 값을 수정하고 싶음 -> WHERE 조건을 기준으로 값을 수정 
    - START TRANSACTION; -> 해당 명령어가 존재한다면, 데이터 수정 후 돌아갈 수 있는 환경을 세팅
    - ROLLBACK ; -> START TRANSACTION; 이 존재한다는 전제하에, 데이터 편집 전 단계로 돌려주는 명령어
    - COMMIT ; -> 해당 명령어를 입력 및 실행하는 순간 임시저장공간에 있던 데이터를 실제 데이터에 적용 
- VIEW
  - VIEW는 실제 테이블을 기반으로 한 "가상 테이블"을 생성하고자 할 때 사용할 수 있는 명령어!!
  - VIEW를 사용하면 복잡한 쿼리문을 단순화 시킬 수 있음!!
  - CREATE VIEW '가상 테이블명' AS 실제 데이터
  - 만약, 동일한 이름으로 가상 테이블이 이미 존재하는 경우, 추가로 생성하고자 할 때는 에러가 발생!!
  - CREATE OR REPLACE VIEW '가상 테이블명' AS 실제 데이터 => 신규로 가상테이블을 생산하거나 또는 덮어쓰기 가능!

- 고급 윈도우 함수

#### SQL 고급문법 
- 윈도우 함수 : SQL 쿼리문 안에서 데이터 집합을 세분화하여 각 부분에 대한 계산을 수행하는 함수
- (*만약 전체데이터가 너무 방대해서, 특정 부분만 추출해서 연산처리를 해야한다면?)
  1) RANK() : 순위를 매기는데, 동일한 값이 있을 경우 같은 순위를부여하고 다음 순위는 건너 뛴다.(1,1,1,4,4,4,7)
  2) DENSE_RANK() : 순위를 매기는데, 동일한 값이 있을 경우, 같은 순위 부여 후 다음 순위를 건너뛰지 않는다 (1,1,1,2,2,2,3)
  3) ROW_NUMBER() : 순위와 상관없이 각 행에 고유한 번호를 부여하는 것 
     > 위 세개의 함수는 공통적으로 OVER 절을 함께 사용한다.
     > ORDER BY 절을 통해서 순위를 매길 기준을 정의한다.

- 윈도우 함수에서 많이 사용되는 구문 :
  1) PARTITION BY : 특정 컬럼을 기준으로 데이터를 부분집합으로 분할할 때 사용
     - COUNT()
     - SUM()
     - AVG()
     - MIN()
     - RANK()
     - DENSE()
     - ROW_NUMBER()
     - LEAD(column, n, default value) : 현재 행을 기준으로 n행 뒤의 값을 가져올 때 사용
     - LAG(column, n, default value) : 현재 행을 기준으로 N행 앞의 값을 가져올 때 사용
     - FIRST_VALUE(): 파티션을 통해 부분집합화 되어진 그룹을 기준으로 첫번째 값을 가져올 때 사용 
     - LAST_VALUE():파티션을 통해 부분집합화 되어진 그룹을 기준으로 마지막 값을 가져올 때 사용
     - 
  2) ORDER BY : 특정 부분집합 내 정렬 기준을 정의하고자 할 때 사용
  3) ROWS : 특정 부분집합의 범위를 설정할 때 사용
     (* 특정 부분집합의 행을 물리적으로 해석)
      - UNBOUNDED PRECEDING : 앞에서 파티션을 통해 설정된 부분집합의 첫 번째 행부터 시작
      - UNBOUNDED FOLLOWING : 앞에서 파티션을 톨해 설정된 부분집합의 마지막번째 행까지
      - CURRENT ROW : 현재 행 포함
      - n PRECEDING / FOLLOWING : n번째 행 앞이나 뒤 
  4) RANGE :정렬하고자 하는 기준값과 동일한 값을 가진 모든 행을 하나의 그룹으로 정의
     (*특정 부분집합 내 동일한 값을 가지고 있는 요소를 하나의 행으로 정의)
  5) SUM(revenue) : 일반적인 집계함수의 기능, 만약 GROUP BY 절과 함께 사용한다면 해당 그룹의 revenue 값을 계산한다고 판단
  6) SUM(revenue) OVER() : 윈도우 함수로서 기능, OVER()는 특정 파티션이 내부에서 정의된 상태라면, 해당 파티션을 기준으로 revenue 값을 집계한다
     하지만 OVER() 안에 특정 파티션이 정의되지 않았다면, 전체 결과 세트에 대해서 SUM 함수를 적용하라는 의미
  7) PERCENT_RANK() : 행의 백분위 순위를 계산 => 0~1 까지, 행의 값을 나눠서 순위를 매길 때 사용 
  8) CUME_DIST() : 행의 누적분포를 계산 => 0~1까지 
  9) NTILE() : 행을 n개의 그룹으로 분할할 때 사용
