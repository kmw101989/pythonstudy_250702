### SQL 학습 전, 반드시 알고가야하는 개념정의
1. 데이터 (Data)
2. 데이터베이스
3. 스키마 (Schema)
4. 테이블 (Table)
5. SQL
    

#### 1. 데이터 (Data) 
- 수집 가능한 값
- 숫자 : 23, 5.9
- 문자 : "david", "2025-07-31"
- 상태 : True , False
- 데이터 그 자체로는 의미가 부족함 -> 구조화가 되어야 데이터는 비로소 가치를 갖게 됨


#### 2. 데이터베이스 (Database = DB)
- 여러 가지 다양한 데이터를 구조적이고 체계적으로 저장 및 관리할 수 있도록 해주는 저장소
- 종류 
   - 관계형 데이터베이스 (RDBMS -> Oracle // MySQL)
   - 비관계형 데이터베이스 (DMBS -> MongoDB)
- 오프라인 가게를 운영하는데 일일 고객 주문 정보, 특정 기간동안 사용한 재고 수량 등

#### 3. 스키마 (Schema)
- 데이터베이스의 구조를 정의한 것
- 데이터베이스를 어떻게 구조화할 것인지를 정의해놓은 일종의 설계도
- 스키마는 데이터베이스 내부에 존재하는 논리적 구조 

#### 4. 테이블 (Table)
- 관계형 데이터베이스 안에 저장된 실제 데이터의 저장 단위
- 구성요소
    - 열 (Column) = 속성 (ex. 고객의 나이, 성별 ,이메일 등)
    - 행 (Row) = 레코드 (ex. 한 명의 고객)


#### 5. SQL (Structured Query Language)
- 관계형 데이터베이스를 정의, 조작,제어 할 수 있도록 도와주는 표준 언어

종류 
 - DDL (Data Definitin Language) : Create , drop ,alter
 - DML (Data Manipulation Lang) : SELECT, INSERT , UPADATE , DELETE
 - DCL (Data Control lang) : Grant, REVOKE

데이터: 가장 하위 개념 . 원초적인 정보

### SQL 중급 단계 
1) FOREIGN KEY (외래 키) <-> PRIMARY KEY (고유 키)

2) HAVING
   - HAVING 절을 활용해서 조건 비교 !!
   - WHERE 절을 활용해서도 가능한데 , 굳이 왜 필요할까?
   - 특정 테이블 내 데이터를 그룹핑하는 GROUP BY를 적용한 상태에서 집계함수가 포함된 조건식을 사용하려고 할 때, 반드시 HAVING 절을 써야 함

3) JOIN
   - 서로 다른 테이블을 하나의 연결 테이블로 활용하고자 할 때 JOIN
   - 복수의 테이블을 관리하는 경우가 빈번 : 각각 테이블 간 연결고리를 FOREIGN KEY 를 만들어놓고 , 해당 테이블간 연결하는 경우가 숙명적으로 많을수밖에 없음 
   - inner join : 서로 다른 테이블간 매칭되는 데이터에 한해서 해당 데이터를 기준으로 연결
   - 서로 다른 테이블을 하나로 연결하기 위해서는 어떤 데이터를 기준으로 연결할 것인가 ? 

   - outer join : 사로 매칭되지 않는 데이터도 모두 찾아옴 !!
       -left OUTER JOIN : 먼저 언급된 (*LEFT data = 부모데이터) 를 기준으로 다른 테이블을 연결 , 자식데이터에 부모데이터와 매칭되지 않은 데이터를 누락시키지 않고                             전부 찾아옴 

    - right OUTER JOIN : 자식데이터를 기준으로 다른 테이블을 연결 , 마찬가지로 매칭되지 않은 데이터들을 누락시키지 않고 전부 찾아옴

4) SubQuery
   - SQL 문 안에 포함되어있는 SQL 문 => 중첩 SQL 문
