```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import re

# -------------------- 기본 설정 --------------------
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--lang=ko_KR")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("start-maximized")
driver = webdriver.Chrome(service=service, options=options)

url = "https://display.wconcept.co.kr/category/women/002002"
driver.get(url)
time.sleep(2)

# -------------------- 카테고리별 크롤링 --------------------
categories = {}

# 카테고리 링크들을 인덱스로 접근
links = driver.find_elements(By.CSS_SELECTOR, "div.swiper-slide")
for idx in range(1, len(links)):  # idx=0은 '전체'라서 제외
    # 매번 다시 elements를 가져와서 stale 방지
    links = driver.find_elements(By.CSS_SELECTOR, "div.swiper-slide")
    link = links[idx]

    index = link.get_attribute("data-value")
    category_name = link.text.strip()
    if not index or not category_name:
        continue

    url2 = f"https://display.wconcept.co.kr/category/women/{index}"
    driver.get(url2)
    time.sleep(2)

    # 페이지 로딩 후 상품 리스트 기다리기
    try:
        WebDriverWait(driver, 10).until(
            EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div.product-item.item.type-all"))
        )
    except:
        print(f"{category_name} 로딩 실패")
        continue

    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1)

    # -------------------- 상품 수집 --------------------
    seen = set()
    products = []

    elements = driver.find_elements(By.CSS_SELECTOR, "div.product-item.item.type-all")
    for element in elements:
        try:
            title = element.find_element(By.CSS_SELECTOR, "span.text.detail").text.strip()
        except:
            title = None

        try:
            rating = element.find_element(By.CSS_SELECTOR, "em.score").text
            rating = float(rating)
        except:
            rating = None

        try:
            review_count = element.find_element(By.CSS_SELECTOR, "span.frpBAf.text.cnt").text
            review_count = review_count.replace("(", "").replace(")", "").replace(",", "")
            review_count = int(review_count)
        except:
            review_count = 0

        item = (title, rating, review_count)

        if item not in seen:
            seen.add(item)
            products.append(item)

    categories[category_name] = products
    print(f"[{category_name}] {len(products)}개 수집 완료")

# -------------------- 종료 --------------------
driver.quit()

# 결과 확인
for cat, prods in categories.items():
    print(f"\n== {cat} ==")
    for p in prods[:3]:
        print("   ", p)
```

```python
import mysql.connector

# 1) DB 연결
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="hithere4961",
)
cur = conn.cursor()

# 2) DB / 테이블 생성
cur.execute("CREATE DATABASE IF NOT EXISTS wconcept CHARACTER SET utf8mb4 COLLATE utf8mb4_unicode_ci;")
cur.execute("USE wconcept;")
cur.execute("""
CREATE TABLE IF NOT EXISTS products (
  id INT AUTO_INCREMENT PRIMARY KEY,
  category VARCHAR(100) NOT NULL,
  title VARCHAR(300) NOT NULL,
  rating DECIMAL(3,1) NULL,
  review_count INT NOT NULL DEFAULT 0,
  UNIQUE KEY uq_cat_title (category, title)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4;
""")

# 3) 딕셔너리를 rows로 변환
rows = []
for cat, prod_list in categories.items():
    for title, rating, review_count in prod_list:
        # 타입 보정(안전)
        r = None if rating is None or str(rating).strip()=="" else float(rating)
        c = 0 if review_count in (None, "") else int(review_count)
        rows.append((cat, title, r, c))

# 4) INSERT (중복이면 업데이트)
sql = """
INSERT INTO products (category, title, rating, review_count)
VALUES (%s, %s, %s, %s)
ON DUPLICATE KEY UPDATE
  rating = VALUES(rating),
  review_count = VALUES(review_count);
"""
cur.executemany(sql, rows)
conn.commit()

cur.close()
conn.close()
print(f"Inserted/Updated {len(rows)} rows.")
```
