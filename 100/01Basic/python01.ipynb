### 2. 다음 뉴스에서 IT 섹션의 기사 제목과 링크를 수집하세요. 


```python
import requests
from bs4 import BeautifulSoup as bs

url ="https://news.daum.net/tech"
headers = {
    "User-Agent":'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36'
} 
res=requests.get(url,headers=headers)
res.encoding = "utf-8"
soup = bs(res.content , "html.parser")


articles = soup.select("a.item_newsheadline2")

for idx , article in enumerate(articles, 1) : 
    title = article.text.strip()
    link = article["href"]
    print(f"{idx}:{title}-{link}\n")
```

### 3. yes24의 경제 베스트셀러 1~10위 도서 제목과 저자  수집

```python
import requests
from bs4 import BeautifulSoup as bs

url ="https://www.yes24.com/product/category/bestseller?categoryNumber=001001025"
headers = {
    "User-Agent":'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36'
} 
res=requests.get(url,headers=headers)
res.encoding = "utf-8"
soup = bs(res.content , "html.parser")

books = soup.select("ul#yesBestList > li")

for i , book in enumerate(books[:10],1):
    title_tag = book.select_one("a.gd_name").text
    auth_tag = book.select_one("span.authPub ").text.strip().strip("저 ")
    print(f"{i}. {auth_tag} - {title_tag}")
```

### 5. 잡코리아에서 '데이터 분석가' 직무 공고 제목 20개 수집 



```python
import requests
from bs4 import BeautifulSoup as bs

q = "데이터 분석가"
url =f"https://www.jobkorea.co.kr/Search/?stext={q}"
headers = {
    "User-Agent":'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36'
} 
res=requests.get(url,headers=headers)
res.encoding = "utf-8"
soup = bs(res.content , "html.parser")

datas = soup.select("div.Flex_display_flex__i0l0hl2.Flex_direction_column__i0l0hl4.h7nnv10")

for idx , data in enumerate(datas,1) :
    comp = data.select_one("span.Typography_variant_size16__344nw26.Typography_weight_regular__344nw2d,Typography_color_gray900__344nw2k")
    titles = data.select_one("span.Flex_display_flex__i0l0hl2.Flex_gap_space4__i0l0hly.Flex_align_center__i0l0hl8")
    print(f"{idx}. {comp.text} - {titles.text}\n")
```

### 6. 5번 문제를 셀레니움으로 크롤링 -> 6개 페이지 크롤링 -> 서울지역만 -> 경력은 1~3년차만 엑셀로 출력 (url링크까지)


```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as ec 
from bs4 import BeautifulSoup
from webdriver_manager.chrome import ChromeDriverManager
import time 
from openpyxl import Workbook
import pandas as pd 
options = Options()
# options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--lang=ko_KR")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service = service  , options = options)

wb = Workbook()
ws = wb.active
ws.append(["기업명", "공고 제목", "링크"])

q = "데이터 분석가"
i = 1
joblists = [] 
jobtuple = ()
for i in range(1,7):
    
    url =f"https://www.jobkorea.co.kr/Search/?stext={q}&Page_No={i}&tabType=recruit&local=I000&careerType=2&careerMin=1&careerMax=3"
    
    driver.get(url)
    
    elements = driver.find_elements(By.CSS_SELECTOR, "div.Flex_display_flex__i0l0hl2.Flex_gap_space28__i0l0hl2a.styles_p_space28__dk46ts8d")
    
    for element in elements :
        cop = element.find_element(By.CSS_SELECTOR,"span.Typography_variant_size16__344nw26.Typography_weight_regular__344nw2d.Typography_color_gray900__344nw2k" ).text
        title = element.find_element(By.CSS_SELECTOR, "span.Typography_variant_size18__344nw25.Typography_weight_medium__344nw2c.Typography_color_gray800__344nw2l").text
        link = element.find_element(By.XPATH, ".//a[@class='sn28bt0']").get_attribute("href")
        # link = element.find_element(By.CSS_SELECTOR, "a.sn28bt0").get_attribute("href")
        jobtuple=(cop,title,link)
        print(jobtuple)
        joblists.append(jobtuple)
    
for job in joblists:
    ws.append(job)
time.sleep(1)
driver.quit()
time.sleep(2)
wb.save("jobkorea_채용공고.xlsx")
print("엑셀 저장 완료!")

df = pd.DataFrame(joblists, columns=["기업명", "공고 제목", "링크"])
df.to_excel("jobkorea_채용공고2.xlsx", index=False)
print("엑셀 저장 완료 (pandas 사용)")
```
