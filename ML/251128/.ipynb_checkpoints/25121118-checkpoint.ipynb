```python
import pickle
import pandas as pd

with open("titanic_step1_feature_change.pickle", "rb") as pickle_filename :
    df = pd.read_pickle(pickle_filename)
```

```python
df.head()
```

```python
# 특정 컬럼값 & 생존.사망 여부 판단
```

```python
title_mapping = {"Mr": 0, "Miss": 1, "Mrs": 2, "Master": 3, "Lady": 4, "Don": 5, "Rev": 6, "Dr": 7,
                 "Major": 8, "Sir": 9, "Col": 10, "Capt": 11, "Jonkheer": 12}

import plotly.express as px

df_initial = df.copy()

df_initial["Outcome"] = df_initial["Survived"].map({0: "Dead", 1: "Survived"})

fig = px.histogram(
    df_initial,
    x="Initial",
    color="Outcome",
    barmode="overlay",
    title="Initial Distribution by Survival Status"
)

fig.update_layout(
    xaxis=dict(
        tickmode="array",
        tickvals=list(title_mapping.values()),
        ticktext=list(title_mapping.keys())
    )
)

fig.show()
```
