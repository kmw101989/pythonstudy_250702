```python
import ssl
import urllib3
from urllib.parse import urlencode
import pandas as pd
import json

# ✅ TLS 1.2 이상 강제
ssl_context = ssl.create_default_context()
ssl_context.options |= ssl.OP_NO_TLSv1 | ssl.OP_NO_TLSv1_1  # TLS1.0, 1.1 비활성화

# ✅ HTTP 연결 생성
http = urllib3.PoolManager(ssl_context=ssl_context)

# ✅ API 기본 URL
base_url = "https://apis.data.go.kr/B551011/GoCamping/basedList"

# ✅ 요청 파라미터 (100개만)
params = {
    'numOfRows': 100,  # ✅ 100개만 가져오기
    'pageNo': 1,
    'MobileOS': 'ETC',
    'MobileApp': 'camping',
    'serviceKey': 'EUkNxC%2BT3jASuqEPBhHgFvYbHdwvvFB5RcniMhZReUDKfgglS7ulOa7oxOzjFDlr5MwDivJX%2FaqrjD881%2BdY1w%3D%3D',
    '_type': 'json'
}

# ✅ URL 조합
url = f"{base_url}?{urlencode(params)}"

# ✅ API 호출
response = http.request("GET", url)

# ✅ 상태 코드 출력
print("응답 코드:", response.status)

# ✅ JSON 파싱
data = json.loads(response.data.decode('utf-8'))

# ✅ 아이템 데이터 추출
items = data['response']['body']['items']['item']

# ✅ pandas DataFrame 변환
df = pd.DataFrame(items)
print(f"총 데이터 개수: {len(df)}")
df.head()

```
