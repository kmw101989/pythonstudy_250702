```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd  # ✅ 추가

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--disable-features=PasswordBreachDetection")
options.add_argument("--disable-popup-blocking")

driver = webdriver.Chrome(service=service, options=options)
url = "https://www.musinsa.com/main/musinsa/recommend?gf=A"
driver.get(url)
time.sleep(2)

items = []
elements = driver.find_elements(By.CSS_SELECTOR, "div.sc-1kwhek5-1.ejsdGf")
for element in elements:
    img = element.find_element(By.CSS_SELECTOR, "img").get_attribute("src")
    title = element.find_element(By.CSS_SELECTOR, 'a').get_attribute("aria-label")
    cell = {
        "title": title,
        "img": img
    }
    items.append(cell)

driver.quit()

# ✅ pandas DataFrame으로 변환
df = pd.DataFrame(items)

# ✅ 엑셀 파일로 저장
output_file = "musinsa_recommend.xlsx"
df.to_excel(output_file, index=False, engine="openpyxl")

print(f"✅ {output_file} 파일이 저장되었습니다!")

```
