```python
# -*- coding: utf-8 -*-
# 네이버 여행지 크롤링 → 영문 변환(googletrans-py) → Nominatim 지오코딩 → CSV 저장
# 작성자: you
# 주의: User-Agent에 본인 연락처를 넣어주세요 (Nominatim 정책)

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import time
import csv
import os
import requests
from googletrans import Translator  # 패키지는 googletrans-py, import 경로는 googletrans

# =========================
# 0) 설정
# =========================
CHROMEDRIVER_PATH = r"C:\Users\user\Downloads\chromedriver\chromedriver.exe"
START_URL = "https://travel.naver.com/overseas?page=1&type=MOST_SEARCHED"
OUTPUT_CSV = "places_with_latlon.csv"

# 반드시 본인 연락 가능한 정보를 넣으세요 (Nominatim 정책: https://operations.osmfoundation.org/policies/nominatim/)
NOMINATIM_UA = "TravelFashionProject/0.5 (contact: your_email@example.com)"  

# 요청 속도(예절)
NOMINATIM_CALL_PAUSE = 1.2

# =========================
# 1) Selenium: 네이버 여행지 크롤링
# =========================
def make_driver():
    options = Options()
    # options.add_argument("--headless=new")  # 필요 시 활성화
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("lang=ko_KR")
    options.add_argument("--start-maximized")
    options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                         "AppleWebKit/537.36 (KHTML, like Gecko) "
                         "Chrome/138.0.0.0 Safari/537.36")
    options.add_argument("--disable-features=PasswordBreachDetection")
    options.add_argument("--disable-popup-blocking")
    service = Service(CHROMEDRIVER_PATH)
    return webdriver.Chrome(service=service, options=options)

def crawl_places():
    """
    반환: [(place, country), ...]
    place: 도시/지역명, country: 국가명 (둘 다 한글일 수 있음)
    """
    driver = make_driver()
    places = []
    try:
        driver.get(START_URL)
        wait = WebDriverWait(driver, 10)
        wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div.item_SeasonalItem__aeQTq")))

        # 카드 더 로드가 필요하면 추가 스크롤
        driver.execute_script("window.scrollTo(0, 2000);")
        time.sleep(1.2)

        elements = driver.find_elements(By.CSS_SELECTOR, "div.item_SeasonalItem__aeQTq")

        for element in elements:
            # place (도시/지역명): 없으면 스킵
            try:
                place = element.find_element(By.CSS_SELECTOR, "span.item_name__tWFCy").text.strip()
            except:
                continue

            # country (국가명): span → 실패시 i도 시도, 그래도 없으면 place로 폴백
            try:
                country = element.find_element(
                    By.CSS_SELECTOR, "span.item_area__6nEmE, i.item_area__6nEmE"
                ).text.strip()
                if not country:
                    country = place
            except:
                country = place

            places.append((place, country))

        return places

    finally:
        try:
            driver.quit()
        except:
            pass

# =========================
# 2) 한→영 변환 (googletrans-py)
# =========================
translator = Translator()

# 흔한 국가는 수동 매핑(정확도↑, 번역 비용↓)
COUNTRY_MAP = {
    "한국": "South Korea", "대한민국": "South Korea",
    "일본": "Japan",
    "미국": "United States",
    "영국": "United Kingdom",
    "태국": "Thailand",
    "베트남": "Vietnam",
    "프랑스": "France",
    "스페인": "Spain",
    "이탈리아": "Italy",
    "독일": "Germany",
}

# 자주 나오는 지명도 일부 매핑
PLACE_MAP = {
    "오키나와": "Okinawa",
    "푸켓": "Phuket",
    "런던": "London",
    # 필요 시 추가
}

def ko2en(text: str) -> str:
    if not text:
        return text
    try:
        return translator.translate(text, src="ko", dest="en").text
    except Exception:
        # 번역 실패 시 원문 반환(후단에서 폴백 지오코딩 시도)
        return text

def normalize_to_en(place_ko: str, country_ko: str) -> tuple[str, str]:
    place_en = PLACE_MAP.get(place_ko.strip(), "") or ko2en(place_ko).strip()
    country_en = COUNTRY_MAP.get(country_ko.strip(), "") or ko2en(country_ko).strip()
    return place_en, country_en

# =========================
# 3) 지오코딩 (Nominatim + ISO 제한 + 폴백)
# =========================
ISO2 = {
    "japan": "jp",
    "south korea": "kr",
    "united states": "us",
    "united kingdom": "gb",
    "thailand": "th",
    "vietnam": "vn",
    "france": "fr",
    "spain": "es",
    "italy": "it",
    "germany": "de",
}

def _nominatim_once(query: str, iso2: str | None, debug: bool = False):
    url = "https://nominatim.openstreetmap.org/search"
    params = {"q": query, "format": "json", "limit": 1}
    if iso2:
        params["countrycodes"] = iso2
    headers = {
        "User-Agent": NOMINATIM_UA,
        "Accept-Language": "en",
    }
    r = requests.get(url, params=params, headers=headers, timeout=12)
    if debug:
        print(f"GET {r.url} -> {r.status_code}")
    if r.status_code != 200:
        return None
    data = r.json()
    if not data:
        return None
    try:
        return float(data[0]["lat"]), float(data[0]["lon"])
    except Exception:
        return None

def geocode_nominatim(place_en: str, country_en: str, retries: int = 2, pause: float = NOMINATIM_CALL_PAUSE, debug: bool = False):
    country_norm = (country_en or "").strip().lower()
    iso2 = ISO2.get(country_norm)

    # 1) place + country (ISO 제한)
    for _ in range(retries):
        res = _nominatim_once(f"{place_en}, {country_en}".strip().strip(","), iso2, debug=debug)
        if res:
            return res
        time.sleep(pause)

    # 2) place만 (ISO 유지)
    for _ in range(retries):
        res = _nominatim_once(place_en, iso2, debug=debug)
        if res:
            return res
        time.sleep(pause)

    # 3) 일본 특수 폴백(현/섬)
    if country_norm == "japan":
        for q in (f"{place_en} Prefecture", f"{place_en} Island"):
            res = _nominatim_once(q, "jp", debug=debug)
            if res:
                return res
            time.sleep(pause)

    # 4) 마지막 폴백: ISO 없이 place만
    for _ in range(retries):
        res = _nominatim_once(place_en, None, debug=debug)
        if res:
            return res
        time.sleep(pause)

    return None, None

# =========================
# 4) 실행 파이프라인
# =========================
def main(debug: bool = False):
    print("[1/4] 네이버 여행지 수집 중…")
    places = crawl_places()
    if not places:
        print("수집된 여행지 없음. 셀렉터 변경 또는 스크롤/대기 조정 필요.")
        return

    print(f"수집 개수: {len(places)} (예: {places[:5]})")

    print("[2/4] 한→영 변환 중…")
    enriched = []
    for place_ko, country_ko in places:
        place_en, country_en = normalize_to_en(place_ko, country_ko)
        enriched.append((place_ko, country_ko, place_en, country_en))

    print("[3/4] 지오코딩(Nominatim) 중…")
    rows = []
    for (place_ko, country_ko, place_en, country_en) in enriched:
        lat, lon = geocode_nominatim(place_en, country_en, debug=debug)
        rows.append({
            "place_ko": place_ko,
            "country_ko": country_ko,
            "place_en": place_en,
            "country_en": country_en,
            "lat": lat,
            "lon": lon,
        })
        time.sleep(NOMINATIM_CALL_PAUSE)  # 예절 준수

    print("[4/4] 결과 저장 중…")
    save_to_csv(rows, OUTPUT_CSV)

    print(f"완료! → {OUTPUT_CSV}")
    # 샘플 출력
    for r in rows[:10]:
        print(r)

def save_to_csv(rows, path):
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    with open(path, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.DictWriter(f, fieldnames=["place_ko", "country_ko", "place_en", "country_en", "lat", "lon"])
        writer.writeheader()
        for r in rows:
            writer.writerow(r)

# =========================
# entry
# =========================
if __name__ == "__main__":
    # debug=True로 하면 각 지오코딩 요청 URL/status를 출력합니다.
    main(debug=False)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
import pyautogui

# === 추가된 모듈 ===
import requests
from googletrans import Translator  # 패키지는 googletrans-py, import는 googletrans
import csv, os

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")  # 경로 확인
options = Options()

# options.add_argument("--headless=new")  # 최신 Chrome에 맞는 headless 모드
options.add_argument("--disable-gpu")   # GPU 비활성화
options.add_argument("--no-sandbox")    # 샌드박스 비활성화 (권한 문제 회피)
options.add_argument("--disable-dev-shm-usage")  # 메모리 공유 문제 해결
options.add_argument("lang=ko_KR")
options.add_argument("--start-maximized")
options.add_argument("--User-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--disable-features=PasswordBreachDetection")
options.add_argument("--disable-popup-blocking")

driver = webdriver.Chrome(service=service, options=options)

# -------------------------
# (1) 네이버 여행지 수집 (원본 구조 최대한 유지)
# -------------------------
places = [] 
for i in range(16): 
    url = f"https://travel.naver.com/overseas?page={i}&type=MOST_SEARCHED"
    driver.get(url)
    time.sleep(2) 
    driver.execute_script("window.scrollTo(0,1980);")
    
    elements = driver.find_elements(By.CSS_SELECTOR , "div.item_SeasonalItem__aeQTq")
    for element in elements:
        # place (도시/지역명): 없으면 스킵
        try:
            place = element.find_element(By.CSS_SELECTOR, "span.item_name__tWFCy").text.strip()
        except:
            continue

        # country (국가명): span → 실패시 i도 함께 시도, 그래도 없으면 place로 폴백
        try:
            country = element.find_element(
                By.CSS_SELECTOR, "span.item_area__6nEmE, i.item_area__6nEmE"
            ).text.strip()
            if not country:
                country = place
        except:
            country = place

        places.append((place, country)) 

for place in places: 
    print(place) 

time.sleep(1)
driver.quit()

# -------------------------
# (2) 한→영 변환 + (3) Nominatim 지오코딩
# -------------------------

# 번역기
translator = Translator()

# 자주 나오는 국가는 수동 매핑(정확도↑, 속도/호출↓)
COUNTRY_MAP = {
    "한국": "South Korea", "대한민국": "South Korea",
    "일본": "Japan",
    "미국": "United States",
    "영국": "United Kingdom",
    "태국": "Thailand",
    "베트남": "Vietnam",
    "프랑스": "France",
    "스페인": "Spain",
    "이탈리아": "Italy",
    "독일": "Germany",
}

# 빈번 지명 매핑(필요시 추가)
PLACE_MAP = {
    "오키나와": "Okinawa",
    "푸켓": "Phuket",
    "런던": "London",
}

def ko2en(text: str) -> str:
    if not text:
        return text
    try:
        return translator.translate(text, src="ko", dest="en").text
    except Exception:
        return text  # 실패 시 원문 유지(뒤에서 폴백 지오코딩 시도)

def normalize_to_en(place_ko: str, country_ko: str) -> tuple[str, str]:
    place_en = PLACE_MAP.get(place_ko.strip(), "") or ko2en(place_ko).strip()
    country_en = COUNTRY_MAP.get(country_ko.strip(), "") or ko2en(country_ko).strip()
    return place_en, country_en

# Nominatim 설정
NOMINATIM_UA = "TravelFashionProject/0.6 (contact: your_email@example.com)"  # ← 본인 이메일/연락처 꼭 넣기
NOMINATIM_CALL_PAUSE = 1.2

ISO2 = {
    "japan": "jp",
    "south korea": "kr",
    "united states": "us",
    "united kingdom": "gb",
    "thailand": "th",
    "vietnam": "vn",
    "france": "fr",
    "spain": "es",
    "italy": "it",
    "germany": "de",
}

def _nominatim_once(query: str, iso2: str | None, debug: bool = False):
    url = "https://nominatim.openstreetmap.org/search"
    params = {"q": query, "format": "json", "limit": 1}
    if iso2:
        params["countrycodes"] = iso2
    headers = {"User-Agent": NOMINATIM_UA, "Accept-Language": "en"}
    r = requests.get(url, params=params, headers=headers, timeout=12)
    if debug:
        print(f"GET {r.url} -> {r.status_code}")
    if r.status_code != 200:
        return None
    try:
        data = r.json()
    except Exception:
        return None
    if not data:
        return None
    try:
        return float(data[0]["lat"]), float(data[0]["lon"])
    except Exception:
        return None

def geocode_nominatim(place_en: str, country_en: str, retries: int = 2, pause: float = NOMINATIM_CALL_PAUSE, debug: bool = False):
    country_norm = (country_en or "").strip().lower()
    iso2 = ISO2.get(country_norm)

    # 1) place + country (ISO 제한)
    for _ in range(retries):
        res = _nominatim_once(f"{place_en}, {country_en}".strip().strip(","), iso2, debug=debug)
        if res:
            return res
        time.sleep(pause)

    # 2) place만 (ISO 유지)
    for _ in range(retries):
        res = _nominatim_once(place_en, iso2, debug=debug)
        if res:
            return res
        time.sleep(pause)

    # 3) 일본 특수 폴백(현/섬)
    if country_norm == "japan":
        for q in (f"{place_en} Prefecture", f"{place_en} Island"):
            res = _nominatim_once(q, "jp", debug=debug)
            if res:
                return res
            time.sleep(pause)

    # 4) 마지막 폴백: ISO 없이 place만
    for _ in range(retries):
        res = _nominatim_once(place_en, None, debug=debug)
        if res:
            return res
        time.sleep(pause)

    return None, None

# -------------------------
# (4) 변환+지오코딩 실행 및 출력/CSV 저장
# -------------------------

# (선택) 중복 제거: 같은 (place,country) 반복 줄이기
# places = list(dict.fromkeys(places))  # 순서 유지한 중복제거 (원하면 주석 해제)

results = []
for place_ko, country_ko in places:
    place_en, country_en = normalize_to_en(place_ko, country_ko)
    lat, lon = geocode_nominatim(place_en, country_en, debug=False)
    if lat is None:  # country가 방해될 때는 place만으로 재시도
        lat, lon = geocode_nominatim(place_en, "", debug=False)
    results.append({
        "place_ko": place_ko,
        "country_ko": country_ko,
        "place_en": place_en,
        "country_en": country_en,
        "lat": lat,
        "lon": lon
    })
    time.sleep(NOMINATIM_CALL_PAUSE)  # 요청 간격 예절

# 확인 출력
for row in results[:30]:
    print(row)

# CSV 저장 (원하면 파일명 변경)
OUTPUT_CSV = "naver_travel_places_latlon.csv"
with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(f, fieldnames=["place_ko","country_ko","place_en","country_en","lat","lon"])
    writer.writeheader()
    writer.writerows(results)

print(f"\n저장 완료 → {OUTPUT_CSV} (총 {len(results)}건)")
```

```python
# -*- coding: utf-8 -*-
# CSV(여행지명, lat, lon, place_en 등) → Open-Meteo로 "오늘" 날씨만 받아 컬럼 추가
# 같은 파일명으로 저장(덮어쓰기)

import time, requests, pandas as pd, numpy as np

# ===== 0) 설정 =====
CSV_PATH = r"naver_travel_places_latlon.csv"  # ← 너의 CSV 경로로
TIMEZONE  = "Asia/Seoul"               # 오늘 기준 타임존
PAUSE_SEC = 0.6                         # API 매너
RETRIES   = 2

ENDPOINT = "https://api.open-meteo.com/v1/forecast"
DAILY_VARS = [
    "temperature_2m_max",
    "temperature_2m_min",
    "relative_humidity_2m_mean",
    "precipitation_sum",
    "uv_index_max",
    "wind_speed_10m_max",
]

# ===== 1) CSV 컬럼 자동 인식 =====
def pick_col(cols, cands):
    low = {c.lower(): c for c in cols}
    for name in cands:
        if name in low:
            return low[name]
    return None

def load_csv(path):
    df = pd.read_csv(path)
    lat_col = pick_col(df.columns, ["lat","latitude","위도"])
    lon_col = pick_col(df.columns, ["lon","lng","longitude","경도"])
    if not lat_col or not lon_col:
        raise ValueError("CSV에 위도/경도 컬럼(lat/lon)이 필요합니다.")
    df["_lat"] = pd.to_numeric(df[lat_col], errors="coerce")
    df["_lon"] = pd.to_numeric(df[lon_col], errors="coerce")
    # 유효 좌표만 대상으로 계산, 나중에 원본에 병합
    ok = (
        df["_lat"].between(-90, 90) &
        df["_lon"].between(-180, 180) &
        ~((df["_lat"] == 0) & (df["_lon"] == 0))
    )
    return df, df.loc[ok].copy(), lat_col, lon_col

# ===== 2) Open-Meteo: 오늘(하루치)만 =====
def fetch_today(lat, lon, retries=RETRIES, pause=PAUSE_SEC):
    params = {
        "latitude": float(lat),
        "longitude": float(lon),
        "timezone": TIMEZONE,
        "daily": ",".join(DAILY_VARS),
        "forecast_days": 1,   # ← 오늘만
        "past_days": 0
    }
    attempt = 0
    while True:
        try:
            r = requests.get(ENDPOINT, params=params, timeout=20)
            if r.status_code == 200:
                data = r.json()
                d = data.get("daily") or {}
                # daily.time[0]에 오늘(타임존 기준) 날짜가 들어옴
                out = {
                    "date": (d.get("time") or [None])[0],
                    "tmax_c": _first(d.get("temperature_2m_max")),
                    "tmin_c": _first(d.get("temperature_2m_min")),
                    "rh_mean_pct": _first(d.get("relative_humidity_2m_mean")),
                    "precip_mm": _first(d.get("precipitation_sum")),
                    "uv_max": _first(d.get("uv_index_max")),
                    "wind_max_ms": _first(d.get("wind_speed_10m_max")),
                }
                time.sleep(pause)
                return out
        except Exception:
            pass
        attempt += 1
        if attempt > retries:
            return {
                "date": None,
                "tmax_c": np.nan, "tmin_c": np.nan,
                "rh_mean_pct": np.nan, "precip_mm": np.nan,
                "uv_max": np.nan, "wind_max_ms": np.nan
            }
        time.sleep(pause)

def _first(arr):
    if isinstance(arr, (list, tuple)) and arr:
        return arr[0]
    return np.nan

# ===== 3) 당일 태그(간단 룰) =====
def today_tags(tmean=None, tmax=None, tmin=None, precip=None, rh=None, uv=None, wind=None):
    tags = set()
    # 온도: mean 없으면 (tmax+tmin)/2 사용
    if tmean is None or pd.isna(tmean):
        if tmax is not None and tmin is not None and not pd.isna(tmax) and not pd.isna(tmin):
            tmean = (tmax + tmin) / 2.0

    if tmean is not None and not pd.isna(tmean):
        if tmean >= 28: tags |= {"ultra-light","linen","mesh","cooling"}
        elif tmean >= 24: tags |= {"lightweight","short-sleeve","breathable"}
        elif tmean >= 18: tags |= {"transitional","layering"}
        elif tmean >= 10: tags |= {"mid-layer","windbreaker"}
        else: tags |= {"insulated","thermal","down"}

    # 강수(일 합계, mm)
    if precip is not None and not pd.isna(precip):
        if precip >= 20: tags |= {"waterproof-jacket","packable-umbrella","quick-dry"}  # 강한 비
        elif precip >= 5: tags |= {"water-resistant","dry-fast"}

    # 습도(%)
    if rh is not None and not pd.isna(rh) and rh >= 75:
        tags |= {"anti-chafe","quick-dry","deodorizing"}

    # UV (WHO 기준: 8+ 매우 높음)
    if uv is not None and not pd.isna(uv):
        if uv >= 8: tags |= {"UPF50+","sun-cap","sunglasses","sleeves"}
        elif uv >= 6: tags |= {"sun-care"}

    # 바람(m/s)
    if wind is not None and not pd.isna(wind) and wind >= 6:
        tags |= {"windbreaker","hooded"}

    return sorted(tags)

# ===== 4) 실행: 오늘 날씨만 붙여 같은 파일에 저장 =====
def main():
    original, valid, lat_col, lon_col = load_csv(CSV_PATH)

    # 결과 컬럼(당일)
    cols_new = ["date_today","tmax_c","tmin_c","rh_mean_pct","precip_mm","uv_max","wind_max_ms","today_tags"]
    for c in cols_new:
        if c not in original.columns:
            original[c] = np.nan

    # 좌표 캐시(중복 호출 방지)
    cache = {}
    for idx, row in valid.iterrows():
        lat, lon = float(row["_lat"]), float(row["_lon"])
        key = (round(lat, 5), round(lon, 5))
        if key not in cache:
            d = fetch_today(lat, lon)
            cache[key] = d
        else:
            d = cache[key]

        tags = today_tags(
            tmean=None,             # 일평균이 없어도 아래에서 tmax/tmin으로 계산
            tmax=d["tmax_c"], tmin=d["tmin_c"],
            precip=d["precip_mm"], rh=d["rh_mean_pct"],
            uv=d["uv_max"], wind=d["wind_max_ms"]
        )

        # 좌표 매칭으로 원본 갱신
        mask = (
            (np.isclose(original[lat_col].astype(float), lat, equal_nan=False)) &
            (np.isclose(original[lon_col].astype(float), lon, equal_nan=False))
        )
        original.loc[mask, "date_today"]   = d["date"]
        original.loc[mask, "tmax_c"]       = d["tmax_c"]
        original.loc[mask, "tmin_c"]       = d["tmin_c"]
        original.loc[mask, "rh_mean_pct"]  = d["rh_mean_pct"]
        original.loc[mask, "precip_mm"]    = d["precip_mm"]
        original.loc[mask, "uv_max"]       = d["uv_max"]
        original.loc[mask, "wind_max_ms"]  = d["wind_max_ms"]
        original.loc[mask, "today_tags"]   = ", ".join(tags) if tags else ""

    # 같은 파일명으로 저장
    original.to_csv(CSV_PATH, index=False, encoding="utf-8-sig")
    print(f"완료: {CSV_PATH} (오늘 날씨 컬럼 갱신)")

if __name__ == "__main__":
    main()
```

```python
# -*- coding: utf-8 -*-
# 기존 CSV(한글 지역/국가, lat/lon) → 영어 지역/국가 컬럼 추가
# lat/lon이 비었거나 이상치면 Nominatim으로 보정
# 저장: "같은 파일명"으로 덮어쓰기

import time, requests, pandas as pd, numpy as np
from pathlib import Path
from googletrans import Translator  # 패키지는 googletrans-py

# ===== 0) 경로 =====
CSV_PATH = r"naver_travel_places_latlon.csv" # ← 네 CSV 경로로 바꿔줘 (같은 이름으로 덮어씀)

# ===== 1) 컬럼 자동 인식 =====
def pick_col(cols, cands):
    low = {c.lower(): c for c in cols}
    for k in cands:
        if k in low: return low[k]
    return None

df = pd.read_csv(CSV_PATH)
col_lat = pick_col(df.columns, ["lat","latitude","위도"])
col_lon = pick_col(df.columns, ["lon","lng","longitude","경도"])
col_place_ko = pick_col(df.columns, ["place_ko","place","여행지","도시","city"])
col_country_ko = pick_col(df.columns, ["country_ko","country","국가"])

if not col_place_ko:
    raise ValueError("한글 지역명 컬럼(place_ko/여행지/도시/PLACE 등)을 찾지 못했습니다.")
if not col_lat or not col_lon:
    # lat/lon이 없다면 이후 지오코딩으로 새로 채움
    col_lat, col_lon = "lat", "lon"
    if "lat" not in df.columns: df["lat"] = np.nan
    if "lon" not in df.columns: df["lon"] = np.nan

# ===== 2) 한→영 변환(번역 + 사전 매핑) =====
translator = Translator()

COUNTRY_MAP = {
    "한국":"South Korea","대한민국":"South Korea",
    "일본":"Japan","미국":"United States","영국":"United Kingdom",
    "태국":"Thailand","베트남":"Vietnam","프랑스":"France",
    "스페인":"Spain","이탈리아":"Italy","독일":"Germany",
}
PLACE_MAP = {
    "오키나와":"Okinawa","푸켓":"Phuket","런던":"London","후쿠오카":"Fukuoka","오사카":"Osaka",
    # 필요시 계속 추가
}

def ko2en(text: str) -> str:
    if pd.isna(text) or str(text).strip() == "":
        return ""
    try:
        return translator.translate(str(text), src="ko", dest="en").text
    except Exception:
        return str(text)

def to_place_en(x): 
    x = str(x).strip()
    return PLACE_MAP.get(x, "") or ko2en(x)

def to_country_en(x):
    x = str(x).strip()
    return COUNTRY_MAP.get(x, "") or ko2en(x)

df["place_en"] = df[col_place_ko].apply(to_place_en)
df["country_en"] = (df[col_country_ko].apply(to_country_en)
                    if col_country_ko else "")

# ===== 3) 위/경도 보정(필요한 행만) =====
# - lat ∉ [-90,90] or lon ∉ [-180,180] or NaN or (0,0) → 재지오코딩
NOMINATIM_UA = "TravelFashionProject/0.7 (contact: youremail@example.com)"  # 본인 정보로 교체
ISO2 = {
    "japan":"jp","south korea":"kr","united states":"us","united kingdom":"gb",
    "thailand":"th","vietnam":"vn","france":"fr","spain":"es","italy":"it","germany":"de",
}

def nominatim_once(query, iso2=None):
    url = "https://nominatim.openstreetmap.org/search"
    params = {"q": query, "format": "json", "limit": 1}
    if iso2: params["countrycodes"] = iso2
    headers = {"User-Agent": NOMINATIM_UA, "Accept-Language": "en"}
    r = requests.get(url, params=params, headers=headers, timeout=12)
    if r.status_code != 200: return None
    data = r.json()
    if not data: return None
    try:
        return float(data[0]["lat"]), float(data[0]["lon"])
    except Exception:
        return None

def geocode(place_en, country_en):
    if not place_en: return None, None
    iso2 = ISO2.get(str(country_en).strip().lower())
    # 1) place + country
    res = nominatim_once(f"{place_en}, {country_en}".strip().strip(","), iso2)
    if res: return res
    time.sleep(1.0)
    # 2) place only (iso 유지)
    res = nominatim_once(place_en, iso2)
    if res: return res
    time.sleep(1.0)
    # 3) 일본 특수(현/섬)
    if str(country_en).strip().lower() == "japan":
        for q in (f"{place_en} Prefecture", f"{place_en} Island"):
            res = nominatim_once(q, "jp")
            if res: return res
            time.sleep(1.0)
    # 4) 마지막: 제한 없이 place
    res = nominatim_once(place_en, None)
    return res if res else (None, None)

# 보정 대상 마스크
lat_num = pd.to_numeric(df[col_lat], errors="coerce")
lon_num = pd.to_numeric(df[col_lon], errors="coerce")
bad = (
    lat_num.isna() | lon_num.isna() |
    (~lat_num.between(-90, 90)) |
    (~lon_num.between(-180, 180)) |
    ((lat_num == 0) & (lon_num == 0))
)

if bad.any():
    print(f"지오코딩 보정 대상: {int(bad.sum())}행")
    for idx in df[bad].index:
        pe, ce = df.at[idx, "place_en"], df.at[idx, "country_en"]
        lat, lon = geocode(pe, ce)
        df.at[idx, col_lat] = lat
        df.at[idx, col_lon] = lon
        time.sleep(1.2)  # Nominatim 예절

# ===== 4) 같은 파일명으로 저장 =====
df.to_csv(CSV_PATH, index=False, encoding="utf-8-sig")
print(f"완료: {CSV_PATH} (place_en, country_en 추가 및 lat/lon 보정)")
```

```python
# -*- coding: utf-8 -*-
# CSV(여행지, lat, lon, place_en …) → OSM 목적지 태그 + Open-Meteo 오늘 날씨
# → today_tags(기후 × 목적지 컨텍스트)까지 생성 후 같은 파일명으로 저장

import time, json, requests, pandas as pd, numpy as np
from collections import Counter

# ========= 설정 =========
CSV_PATH  = r"naver_travel_places_latlon.csv"  # ← 너의 CSV 경로
TIMEZONE  = "Asia/Seoul"                # 오늘 기준 타임존
RADIUS_M  = 3000                        # OSM 반경(m) — 도시 2~3km, 섬/리조트 5~8km도 OK
PAUSE_OSM = 1.2                         # Overpass 예절(최소 1초 권장)
PAUSE_WX  = 0.6                         # Open-Meteo 예절
RETRIES   = 2
UA_OSM    = "TravelFashionProject/OSM/1.0 (contact: youremail@example.com)"  # 연락가능 UA

OVERPASS  = "https://overpass-api.de/api/interpreter"
WX_EP     = "https://api.open-meteo.com/v1/forecast"
DAILY_VARS = [
    "temperature_2m_max","temperature_2m_min",
    "relative_humidity_2m_mean","precipitation_sum",
    "uv_index_max","wind_speed_10m_max"
]

# ========= 공통 유틸 =========
def pick_col(cols, cands):
    low = {c.lower(): c for c in cols}
    for n in cands:
        if n in low: return low[n]
    return None

def load_csv(path):
    df = pd.read_csv(path)
    lat_col = pick_col(df.columns, ["lat","latitude","위도"])
    lon_col = pick_col(df.columns, ["lon","lng","longitude","경도"])
    if not lat_col or not lon_col:
        raise ValueError("CSV에 위도/경도 컬럼(lat/lon)이 필요합니다.")
    # 유효 좌표 마스크
    lat = pd.to_numeric(df[lat_col], errors="coerce")
    lon = pd.to_numeric(df[lon_col], errors="coerce")
    ok = (
        lat.between(-90,90) & lon.between(-180,180) &
        ~((lat==0) & (lon==0))
    )
    return df, lat_col, lon_col, ok

# ========= OSM(Overpass) =========
def overpass_query(lat, lon, radius=RADIUS_M):
    q = f"""
    [out:json][timeout:25];
    (
      node(around:{radius},{lat},{lon})[tourism];
      way(around:{radius},{lat},{lon})[tourism];
      relation(around:{radius},{lat},{lon})[tourism];

      node(around:{radius},{lat},{lon})[natural~"^(beach|coastline|bay|waterfall|peak|volcano)$"];
      way(around:{radius},{lat},{lon})[natural~"^(beach|coastline|bay|waterfall|peak|volcano)$"];
      relation(around:{radius},{lat},{lon})[natural~"^(beach|coastline|bay|waterfall|peak|volcano)$"];

      node(around:{radius},{lat},{lon})[leisure~"^(park|marina|beach_resort|golf_course|stadium)$"];
      way(around:{radius},{lat},{lon})[leisure~"^(park|marina|beach_resort|golf_course|stadium)$"];
      relation(around:{radius},{lat},{lon})[leisure~"^(park|marina|beach_resort|golf_course|stadium)$"];

      node(around:{radius},{lat},{lon})[historic];
      way(around:{radius},{lat},{lon})[historic];
      relation(around:{radius},{lat},{lon})[historic];

      node(around:{radius},{lat},{lon})[sport~"^(surfing|skiing|scuba_diving|windsurfing)$"];
      way(around:{radius},{lat},{lon})[sport~"^(surfing|skiing|scuba_diving|windsurfing)$"];
      relation(around:{radius},{lat},{lon})[sport~"^(surfing|skiing|scuba_diving|windsurfing)$"];

      node(around:{radius},{lat},{lon})[place~"^(island)$"];
      way(around:{radius},{lat},{lon})[place~"^(island)$"];
      relation(around:{radius},{lat},{lon})[place~"^(island)$"];
    );
    out tags;
    """
    r = requests.post(OVERPASS, data=q.encode("utf-8"),
                      headers={"User-Agent": UA_OSM, "Content-Type": "text/plain"},
                      timeout=30)
    r.raise_for_status()
    data = r.json()
    return [el.get("tags", {}) for el in data.get("elements", [])]

def rollup_destination_tags(tags_list):
    c = Counter()
    for t in tags_list:
        n = t.get("natural")
        if n in {"beach","coastline","bay","waterfall","peak","volcano"}: c[n]+=1
        if "tourism" in t:   c[f"tourism:{t['tourism']}"] += 1
        if "leisure" in t:   c[f"leisure:{t['leisure']}"] += 1
        if "historic" in t:  c[f"historic:{t['historic']}"] += 1
        if "sport" in t:     c[f"sport:{t['sport']}"] += 1
        if t.get("place") == "island": c["island"] += 1

    dest = set()
    # 대표 규칙(필요시 튜닝)
    if c["beach"] + c["leisure:beach_resort"] >= 3: dest.add("beach")
    if c["island"] >= 1: dest.add("island")
    if (c["tourism:museum"] + c["historic:yes"] + c["tourism:attraction"]) >= 8: dest.add("culture")
    if (c["peak"] + c["volcano"]) >= 3 or c["leisure:park"] >= 12: dest.add("nature")
    if c["leisure:marina"] >= 2: dest.add("marina")
    if c["sport:surfing"] >= 1 or c["sport:windsurfing"] >= 1: dest.add("surf")
    if c["sport:scuba_diving"] >= 1: dest.add("diving")
    if c["sport:skiing"] >= 1 or c["tourism:ski_resort"] >= 1: dest.add("ski")
    if c["tourism:theme_park"] >= 1: dest.add("theme-park")
    if c["tourism:viewpoint"]  >= 2: dest.add("viewpoint")
    return ", ".join(sorted(dest)), c

# ========= Open-Meteo(오늘 하루) =========
def _first(arr):
    if isinstance(arr, (list, tuple)) and arr:
        return arr[0]
    return np.nan

def fetch_today(lat, lon, pause=PAUSE_WX, retries=RETRIES, backoff=1.5):
    params = {
        "latitude": float(lat),
        "longitude": float(lon),
        "timezone": TIMEZONE,
        "daily": ",".join(DAILY_VARS),
        "forecast_days": 1,
        "past_days": 0
    }
    wait, attempt = pause, 0
    while True:
        try:
            r = requests.get(WX_EP, params=params, timeout=25)
            if r.status_code == 200:
                d = (r.json() or {}).get("daily") or {}
                out = {
                    "date":           (d.get("time") or [None])[0],
                    "tmax_c":         _first(d.get("temperature_2m_max")),
                    "tmin_c":         _first(d.get("temperature_2m_min")),
                    "rh_mean_pct":    _first(d.get("relative_humidity_2m_mean")),
                    "precip_mm":      _first(d.get("precipitation_sum")),
                    "uv_max":         _first(d.get("uv_index_max")),
                    "wind_max_ms":    _first(d.get("wind_speed_10m_max")),
                }
                time.sleep(pause)
                return out
        except Exception:
            pass
        attempt += 1
        if attempt > retries:
            return {"date":None,"tmax_c":np.nan,"tmin_c":np.nan,"rh_mean_pct":np.nan,
                    "precip_mm":np.nan,"uv_max":np.nan,"wind_max_ms":np.nan}
        time.sleep(wait); wait *= backoff

# ========= 기후→스타일(오늘) =========
def today_tags(tmean=None, tmax=None, tmin=None, precip=None, rh=None, uv=None, wind=None):
    # 일평균이 없으면 (tmax+tmin)/2
    if tmean is None or pd.isna(tmean):
        if tmax is not None and tmin is not None and not pd.isna(tmax) and not pd.isna(tmin):
            tmean = (tmax + tmin) / 2.0
    tags = set()
    # 온도
    if tmean is not None and not pd.isna(tmean):
        if tmean >= 28: tags |= {"ultra-light","linen","mesh","cooling"}
        elif tmean >= 24: tags |= {"lightweight","short-sleeve","breathable"}
        elif tmean >= 18: tags |= {"transitional","layering"}
        elif tmean >= 10: tags |= {"mid-layer","windbreaker"}
        else: tags |= {"insulated","thermal","down"}
    # 강수(일합 mm)
    if precip is not None and not pd.isna(precip):
        if precip >= 20: tags |= {"waterproof-jacket","packable-umbrella","quick-dry"}
        elif precip >= 5: tags |= {"water-resistant","dry-fast"}
    # 습도(%)
    if rh is not None and not pd.isna(rh) and rh >= 75:
        tags |= {"anti-chafe","quick-dry","deodorizing"}
    # UV
    if uv is not None and not pd.isna(uv):
        if uv >= 8: tags |= {"UPF50+","sun-cap","sunglasses","sleeves"}
        elif uv >= 6: tags |= {"sun-care"}
    # 바람(m/s)
    if wind is not None and not pd.isna(wind) and wind >= 6:
        tags |= {"windbreaker","hooded"}
    return ", ".join(sorted(tags))

# ========= 목적지 컨텍스트 반영 =========
def _parse_tags_str(s: str):
    if not isinstance(s, str) or not s.strip():
        return set()
    return {t.strip() for t in s.split(",") if t.strip()}

def apply_dest_context(climate_tags, dest_tags_osm, uv=None, precip=None, wind=None):
    # climate_tags: 문자열/리스트/셋 모두 허용
    base = _parse_tags_str(climate_tags) if not isinstance(climate_tags, (list,set,tuple)) else set(climate_tags)
    dest = _parse_tags_str(dest_tags_osm)

    extra = set()
    if {"beach","island"} & dest:
        extra |= {"swimwear","rashguard","water-shoes","sandals"}
        if uv is not None and not pd.isna(uv) and uv >= 6:
            extra.add("UPF50+")
    if "surf" in dest:
        extra |= {"rashguard","boardshorts","water-shoes"}
    if "diving" in dest:
        extra |= {"rashguard","boat-jacket"}
    if "ski" in dest:
        extra |= {"snow-pants","thermal","gloves","beanie","neck-warmer","waterproof-jacket"}
    if "nature" in dest:
        extra |= {"hiking-shoes","trail-pants","sun-cap"}
    if "marina" in dest:
        extra |= {"deck-shoes","light-cardigan"}
    if "culture" in dest:
        extra |= {"walking-shoes","crossbody-bag"}
    if "theme-park" in dest:
        extra |= {"comfortable-sneakers","packable-poncho"}
    if "viewpoint" in dest and wind is not None and not pd.isna(wind) and wind >= 6:
        extra.add("windbreaker")

    # 강수 시 야외형 목적지 보강
    if precip is not None and not pd.isna(precip) and precip >= 10 and ({"beach","theme-park"} & dest):
        extra.add("packable-poncho")

    final = sorted(base | extra)
    return ", ".join(final)

# ========= 메인: 통합 실행 & 저장 =========
def main():
    df, lat_col, lon_col, ok = load_csv(CSV_PATH)

    # 결과 컬럼 준비(없으면 생성)
    for c in ["dest_tags_osm","osm_raw_counts","date_today","tmax_c","tmin_c",
              "rh_mean_pct","precip_mm","uv_max","wind_max_ms","today_tags"]:
        if c not in df.columns: df[c] = np.nan

    # 캐시(좌표 단위)
    osm_cache, wx_cache = {}, {}

    for i, row in df[ok].iterrows():
        lat = float(row[lat_col]); lon = float(row[lon_col])
        key = (round(lat,4), round(lon,4))

        # --- OSM 목적지 태그 ---
        if key not in osm_cache:
            try:
                tags_list = overpass_query(lat, lon, RADIUS_M)
                dest, counts = rollup_destination_tags(tags_list)
                osm_cache[key] = (dest, counts)
                time.sleep(PAUSE_OSM)
            except Exception:
                osm_cache[key] = ("", {})
        dest, counts = osm_cache[key]
        df.at[i, "dest_tags_osm"] = dest
        df.at[i, "osm_raw_counts"] = json.dumps(counts, ensure_ascii=False) if counts else ""

        # --- 오늘 날씨 ---
        if key not in wx_cache:
            wx_cache[key] = fetch_today(lat, lon)
        d = wx_cache[key]
        df.at[i, "date_today"]  = d["date"]
        df.at[i, "tmax_c"]      = d["tmax_c"]
        df.at[i, "tmin_c"]      = d["tmin_c"]
        df.at[i, "rh_mean_pct"] = d["rh_mean_pct"]
        df.at[i, "precip_mm"]   = d["precip_mm"]
        df.at[i, "uv_max"]      = d["uv_max"]
        df.at[i, "wind_max_ms"] = d["wind_max_ms"]

        # --- today_tags = 기후 × 목적지 컨텍스트 ---
        climate_only = today_tags(
            tmax=d["tmax_c"], tmin=d["tmin_c"], precip=d["precip_mm"],
            rh=d["rh_mean_pct"], uv=d["uv_max"], wind=d["wind_max_ms"]
        )
        df.at[i, "today_tags"]  = apply_dest_context(
            climate_only, df.at[i, "dest_tags_osm"],
            uv=d["uv_max"], precip=d["precip_mm"], wind=d["wind_max_ms"]
        )

    # 저장(덮어쓰기)
    df.to_csv(CSV_PATH, index=False, encoding="utf-8-sig")
    print(f"완료: {CSV_PATH} (dest_tags_osm + 오늘 날씨 + today_tags 갱신)")

if __name__ == "__main__":
    main()
```
