#### 문자열 관련 함수 리스트 
- count : 문자열 객체 안에 있는 특정 문자의 개수를 세기
- index : 문자열에 있는 특정 문자의 인덱스 위치를 알려줌
- find : 문자열에 있는 특정 문자의 인덱스를 찾아오긴 하나, 만약 존재하지 않는다면 에러를 발생하지 않고 -1 이 발생
- join : 문자열 사이에 다른 문자를 끼워넣기
- strip : 문자열 앞뒤에 공백 지우기
- lower : 대문자를 소문자로 변경하기
- upper : 소문자를 대문자로 변경하기
- split : 문자열 내 특정 문자를 기준으로 나누기
- replace : 문자열 중 일부를 다른 문자로 바꾸거나 혹은 삭제하기 

```python
data ="dave david"
data.count("")
```

```python
string = "dave ID is dave"
string.index("D")
```

```python
string01 = "Dave ID is dave"
string01.find("d")

if string01.find("x") == -1 :
    print("x는 string01에 존재하지 않음")
```

#### 문자열 "10, 11, 12,13, 14" 해당 값을 리스트 자료형태 [ 10,11,...로 변환 

```python
string = "10,11,12,13,14"
data = string.split(",")
data1 = []

for item in data:
    data1.append(int(item))

print(data1)
```

```python
string = "10,11,12,13,14"
split_string = string.split(",")

for index, split_item in enumerate(split_string) : 
    split_string[index] = int(split_item)

print(split_string)
```

```python
string08 = "david goes to Korea" 
string08.replace("goes","went")

string09 = "(david)" 
string09.replace("()","")

string10 = "(david)" 
string10.replace("(","").replace(")","")

```
